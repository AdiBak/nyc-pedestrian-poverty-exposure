**Step 2:** Begin by creating a chart that displays the average number of crashes per day of the week, including Monday, Tuesday, Wednesday, Thursday, Friday, Saturday, & Sunday.

````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Load the dataset
file_path = "/content/drive/MyDrive/Motor_Vehicle_Collisions_-_Crashes_20260810.csv"
data = pd.read_csv(file_path)

# TODO: Convert 'CRASH DATE' to datetime
data['CRASH DATE'] = pd.to_datetime(data['CRASH DATE'])

# Day of the Week Analysis
# TODO: Add a new column for the day of the week from 'CRASH DATE'
data['Day of Week'] = data['CRASH DATE'].dt.day_name()
day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']

# Group by 'Day of Week' and calculate the average number of crashes per day
average_crashes_per_weekday = data.groupby('Day of Week').size() / data['CRASH DATE'].nunique()

# Plot the average number of crashes
plt.figure(figsize=(10, 6))
# TODO: Plot a bar chart with seaborn
sns.barplot(x=average_crashes_per_weekday.index, y=average_crashes_per_weekday.values, order=day_order)
plt.title('Average Number of Crashes per Day of the Week')
plt.xlabel('Day of the Week')
plt.ylabel('Average Number of crashes')
plt.xticks(rotation=45)
plt.show()
````

**Step 3:** [Stationarity Check](https://otexts.com/fpp2/stationarity.html)! Time series data should be stationary for most forecasting methods to work effectively. Let's check if the series is stationary and apply transformations if necessary, leveraging the [Augmented Dickey-Fuller test](https://www.machinelearningplus.com/time-series/augmented-dickey-fuller-test/).

````python
from statsmodels.tsa.stattools import adfuller

# TODO: Count the number of crashes per day
daily_crashes = data.groupby('CRASH DATE').size()

# Perform Augmented Dickey-Fuller test
# TODO: Use adfuller to test if daily crashes are stationary
adf_test = adfuller(daily_crashes)

# Display the ADF test results
# TODO: Create a pandas Series to nicely display test results
adf_output = pd.Series(adf_test[0:4], index=['Test Statistic','p-value','#Lags Used','Number of Observations Used'])
for key, value in adf_test[4].items():
    adf_output[f'Critical Value ({key})'] = value

# TODO: Print the ADF test output
print(adf_output)
````

[Review this resource for a refresher on test statistics and P-values](https://www.investopedia.com/terms/p/p-value.asp). Then, review the results of the Augmented Dickey-Fuller (ADF) test below:

<font color='blue'>Test Statistic: -1.294

p-value: 0.632</font>

The p-value is greater than the common significance level (0.05), indicating that we *fail to reject the null hypothesis*. Therefore, we can infer that the time series is *not stationary*.
We typically apply differencing to make the series stationary. After differencing, we can recheck for stationarity and then proceed with selecting and fitting an appropriate forecasting model.

**Step 4:** In order to make the series stationary, let's apply differencing. [Review this resource to learn more about Time Series Differencing](https://www.influxdata.com/blog/time-series-differencing-complete-guide-influxdb/). **Then, perform the ADF test on the differenced series, and display the results.**

````python
# TODO: Apply first differencing
daily_crashes_diff = daily_crashes.diff().dropna()

# Perform Augmented Dickey-Fuller test on the differenced series
# TODO: Use adfuller to test if the differenced series is stationary
adf_test_diff = adfuller(daily_crashes_diff)

# Display the ADF test results for the differenced series
# TODO: Create a pandas Series to nicely display test results
adf_output_diff = pd.Series(adf_test_diff[0:4], index=['Test Statistic','p-value','#Lags Used','Number of Observations Used'])
for key, value in adf_test_diff[4].items():
    adf_output_diff[f'Critical Value ({key})'] = value

# TODO: Print the ADF test output for the differenced series
print(adf_output_diff)
````

The results of the Augmented Dickey-Fuller test on the differenced series are:

Test Statistic: Approximately -19.85

p-value: Approximately 0

The p-value is significantly lower than common significance level (0.05).

````python
# Imports and settings
import io
import os
import time
import urllib.request

import geopandas as gpd
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import requests
import statsmodels
import statsmodels.api as sm
import statsmodels.formula.api as smf
from google.colab import userdata

# Define study windows and population thresholds
STUDY_START, STUDY_END = 2019, 2024
HEADLINE_YEARS = list(range(2019, 2024))        # matches the ACS 2019-2023 window
EARLY_YEARS, LATE_YEARS = [2019, 2020, 2021], [2022, 2023, 2024]
MIN_POP_ZIP = 5000                              # resident floor for a ZIP to enter the analysis
MIN_POP_TRACT = 1000                            # resident floor for a tract
SNAP_FT = 1640                                  # snap radius for points on bridges or water (about 500 m)
NYC_COUNTIES = ['005', '047', '061', '081', '085']
HWY_WORDS = ['EXPRESSWAY', 'PARKWAY', 'BRIDGE', 'HIGHWAY', 'TUNNEL', 'THRUWAY', 'TURNPIKE', 'FDR', 'BQE']
CONTROLS = ['Walk_Share', 'Transit_Share', 'Commercial_Share']

# Model specifications for later evaluation
SPECS = [('poverty only', []),
         ('+ walking, transit', CONTROLS[:2]),
         ('+ walking, transit, commercial', CONTROLS)]
OUTCOMES = [('Ped_Killed', 'Deaths'), ('Ped_Injured', 'Injuries'), ('Crashes', 'All crashes')]

# Create output directory and configure pandas display
OUT_DIR = '/content/results'
os.makedirs(OUT_DIR, exist_ok=True)
API_KEY = userdata.get('CENSUS_API_KEY')
pd.set_option('display.width', 250, 'display.max_colwidth', 45)

print(f"pandas {pd.__version__}, statsmodels {statsmodels.__version__}, geopandas {gpd.__version__}")
````

````python
# Crash data window setup
crashes = data.copy()
# Parse crash dates to datetime format
crashes['CRASH DATE'] = pd.to_datetime(crashes['CRASH DATE'])
print("Latest crash date in file:", crashes['CRASH DATE'].max().date())
# Filter records to our target study timeframe
crashes['YEAR'] = crashes['CRASH DATE'].dt.year
crashes = crashes[crashes['YEAR'].between(STUDY_START, STUDY_END)].copy()

per_year = crashes.groupby('YEAR').size()
print(per_year)
if crashes['CRASH DATE'].max() < pd.Timestamp(f'{STUDY_END}-12-01'):
    print(f"WARNING: {STUDY_END} looks incomplete.")

# Safely convert ZIP to numeric, coercing errors to NaN
crashes['ZIP_REPORTED'] = pd.to_numeric(crashes['ZIP CODE'], errors='coerce')
# Generate a heuristic road type flag by checking for highway-related keywords
street_text = (crashes['ON STREET NAME'].fillna('') + ' ' + crashes['CROSS STREET NAME'].fillna('')).str.upper()
crashes['is_highway'] = street_text.str.contains('|'.join(HWY_WORDS), regex=True)   # approximate road type flag

fatal = crashes[crashes['NUMBER OF PEDESTRIANS KILLED'] > 0]
print(f"Fatal pedestrian crashes missing a ZIP: {fatal['ZIP_REPORTED'].isna().mean():.1%}")
print(f"Highway flagged, with a ZIP: {fatal.loc[fatal['ZIP_REPORTED'].notna(), 'is_highway'].mean():.1%}, "
      f"without a ZIP: {fatal.loc[fatal['ZIP_REPORTED'].isna(), 'is_highway'].mean():.1%}")
````

````python
# Boundaries, tract assignment and ZIP recovery

def download(url, path):
    """Helper to fetch geographic files locally."""
    if not os.path.exists(path):
        urllib.request.urlretrieve(url, path)
    return path

# Load ZCTA and Census Tract shapefiles
zcta = gpd.read_file(download(
    "https://www2.census.gov/geo/tiger/GENZ2020/shp/cb_2020_us_zcta520_500k.zip", "/content/zcta2020.zip"))
zcode = [c for c in zcta.columns if c.startswith('ZCTA5CE')][0]
tracts = gpd.read_file(download(
    "https://www2.census.gov/geo/tiger/GENZ2020/shp/cb_2020_36_tract_500k.zip", "/content/ny_tracts2020.zip"))
# Filter to the five NYC counties and reproject to NY State Plane (feet) for accurate distance measurements
tracts = tracts[tracts['COUNTYFP'].isin(NYC_COUNTIES)][['GEOID', 'geometry']].to_crs(2263)
print(f"NYC tracts: {len(tracts)}")

# Only consider crashes with plausible geographic coordinates
valid = crashes['LATITUDE'].between(40.4, 41.0) & crashes['LONGITUDE'].between(-74.3, -73.6)
pts = gpd.GeoDataFrame(
    crashes.loc[valid, []],
    geometry=gpd.points_from_xy(crashes.loc[valid, 'LONGITUDE'], crashes.loc[valid, 'LATITUDE']),
    crs=4326).to_crs(2263)

# Every crash with coordinates gets a tract. Points inside match at distance 0, others snap within SNAP_FT.
jt = gpd.sjoin_nearest(pts, tracts, how='left', max_distance=SNAP_FT)
jt = jt[~jt.index.duplicated(keep='first')]
crashes['TRACT'] = jt['GEOID'].reindex(crashes.index)

# Crashes without a reported ZIP get one from the ZCTA boundaries
xmin, ymin, xmax, ymax = pts.total_bounds
zcta_nyc = zcta.to_crs(2263).cx[xmin - 5000:xmax + 5000, ymin - 5000:ymax + 5000]
pts_zip = pts[crashes.loc[pts.index, 'ZIP_REPORTED'].isna()]
jz = gpd.sjoin_nearest(pts_zip, zcta_nyc[[zcode, 'geometry']], how='left', max_distance=SNAP_FT)
jz = jz[~jz.index.duplicated(keep='first')]
crashes['ZIP_RECOVERED'] = pd.to_numeric(jz[zcode], errors='coerce').reindex(crashes.index)

# Store different ZIP scenarios for our sensitivity analysis
crashes['ZIP_ALL'] = crashes['ZIP_REPORTED'].fillna(crashes['ZIP_RECOVERED'])
crashes['ZIP_LOCAL'] = crashes['ZIP_REPORTED'].fillna(crashes['ZIP_RECOVERED'].where(~crashes['is_highway']))

# Cluster id for each tract is the ZIP that contains its centroid. Important for clustering standard errors.
tr_c = tracts.copy()
tr_c['geometry'] = tr_c.geometry.centroid
cl = gpd.sjoin_nearest(tr_c, zcta_nyc[[zcode, 'geometry']], how='left')
tract_cluster = cl[~cl.index.duplicated(keep='first')][['GEOID', zcode]].rename(columns={zcode: 'ZCTA_cluster'})

head = crashes[crashes['YEAR'].isin(HEADLINE_YEARS)]
deaths = head['NUMBER OF PEDESTRIANS KILLED']
print("Share of pedestrian deaths placed, headline window")
for col in ['ZIP_REPORTED', 'ZIP_ALL', 'ZIP_LOCAL', 'TRACT']:
    print(f"  {col}: {deaths[head[col].notna()].sum() / deaths.sum():.1%}")
````

````python
# Neighborhood data from the Census API and PLUTO
ACS_BASE = "https://api.census.gov/data/2023/acs/acs5/subject"
# 2019-2023 5-year estimates
ACS_VARS = {'S1701_C01_001E': 'Total_Population', 'S1701_C03_001E': 'Poverty_Rate_Percent',
            'S0801_C01_010E': 'Walk_Share', 'S0801_C01_009E': 'Transit_Share'}

def fetch_acs(geo_clause):
    url = f"{ACS_BASE}?get=NAME,{','.join(ACS_VARS)}&{geo_clause}&key={API_KEY}"
    r = requests.get(url, timeout=180)
    r.raise_for_status()
    rows = r.json()
    d = pd.DataFrame(rows[1:], columns=rows[0]).rename(columns=ACS_VARS)
    for c in ACS_VARS.values():
        d[c] = pd.to_numeric(d[c], errors='coerce')
        d.loc[d[c] < 0, c] = np.nan                     # suppressed values arrive as large negatives
    return d

zip_acs = fetch_acs("for=zip%20code%20tabulation%20area:*")
zip_acs['ZIP'] = pd.to_numeric(zip_acs['zip code tabulation area'], errors='coerce')
zip_acs = zip_acs.dropna(subset=['ZIP']).astype({'ZIP': int})

tract_acs = fetch_acs(f"for=tract:*&in=state:36&in=county:{','.join(NYC_COUNTIES)}")
tract_acs['GEOID'] = tract_acs['state'] + tract_acs['county'] + tract_acs['tract']

PLUTO_URL = "https://data.cityofnewyork.us/resource/64uk-42ks.csv"

def get_with_retry(params, tries=5):
    for i in range(tries):
        r = requests.get(PLUTO_URL, params=params, timeout=180)
        if r.status_code == 200:
            return r
        print(f"PLUTO attempt {i + 1}: HTTP {r.status_code}, waiting {5 * 2 ** i}s")
        time.sleep(5 * 2 ** i)
    r.raise_for_status()

def pluto_share(group):
    r = get_with_retry({'$select': f'{group}, sum(comarea) as com_area, sum(resarea) as res_area',
                        '$group': group, '$limit': 5000})
    p = pd.read_csv(io.StringIO(r.text), dtype={group: str})
    p['Commercial_Share'] = p['com_area'] / (p['com_area'] + p['res_area']) * 100
    return p

pz = pluto_share('zipcode')
pz['ZIP'] = pd.to_numeric(pz['zipcode'], errors='coerce')
pluto_zip = pz.dropna(subset=['ZIP']).astype({'ZIP': int})[['ZIP', 'Commercial_Share']]

pt = pluto_share('bct2020')
pt['bct2020'] = pt['bct2020'].astype(str).str.replace(r'\.0$', '', regex=True).str.zfill(7)
boro_to_county = {'1': '061', '2': '005', '3': '047', '4': '081', '5': '085'}
pt = pt[pt['bct2020'].str[0].isin(boro_to_county)].copy()
pt['GEOID'] = '36' + pt['bct2020'].str[0].map(boro_to_county) + pt['bct2020'].str[1:]
pluto_tract = pt[['GEOID', 'Commercial_Share']]
print(f"ZIPs with ACS data: {len(zip_acs)}, NYC tracts with ACS data: {len(tract_acs)}, "
      f"PLUTO ZIPs: {len(pluto_zip)}, PLUTO tracts: {len(pluto_tract)}")
````

````python
# Table builders and model helpers
EVENTS = dict(Ped_Killed=('NUMBER OF PEDESTRIANS KILLED', 'sum'),
              Ped_Injured=('NUMBER OF PEDESTRIANS INJURED', 'sum'),
              Crashes=('CRASH DATE', 'size'))
NEED = ['Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share', 'Commercial_Share']

def zip_table(years, zip_col, min_pop=MIN_POP_ZIP):
    """Aggregate crash and census data at the ZIP code level."""
    d = crashes[crashes['YEAR'].isin(years)].dropna(subset=[zip_col]).copy()
    d[zip_col] = d[zip_col].astype(int)
    agg = d.groupby(zip_col).agg(**EVENTS).reset_index().rename(columns={zip_col: 'ZIP'})
    # Merge in Census and PLUTO features
    m = (agg.merge(zip_acs[['ZIP', 'Total_Population', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share']],
                   on='ZIP', how='inner')
            .merge(pluto_zip, on='ZIP', how='left'))
    m = m[m['Total_Population'] >= min_pop].dropna(subset=NEED).reset_index(drop=True)
    m['n_years'] = len(years)
    return m

def tract_table(years, scenario='local', min_pop=MIN_POP_TRACT):
    """Aggregate crash and census data at the Census Tract level."""
    d = crashes[crashes['YEAR'].isin(years)].dropna(subset=['TRACT'])
    if scenario == 'local':
        d = d[~d['is_highway']]                         # drop highway flagged crashes entirely
    agg = d.groupby('TRACT').agg(**EVENTS).reset_index().rename(columns={'TRACT': 'GEOID'})
    m = (tract_acs[['GEOID', 'Total_Population', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share']]
         .merge(agg, on='GEOID', how='left')            # tracts with no crashes stay in as zeros
         .merge(pluto_tract, on='GEOID', how='left')
         .merge(tract_cluster, on='GEOID', how='left'))
    m[list(EVENTS)] = m[list(EVENTS)].fillna(0)
    m = m[m['Total_Population'] >= min_pop].dropna(subset=NEED + ['ZCTA_cluster']).reset_index(drop=True)
    m['n_years'] = len(years)
    return m

def fit_nb(m, outcome, controls, groups=None):
    """Negative binomial regression model with population exposure. Pass groups to cluster the standard errors."""
    rhs = ' + '.join(['Poverty_Rate_Percent'] + controls)
    model = smf.negativebinomial(f'{outcome} ~ {rhs}', data=m, exposure=m['Total_Population'] * m['n_years'])
    if groups is None:
        return model.fit(disp=0, maxiter=500)
    return model.fit(disp=0, maxiter=500, cov_type='cluster', cov_kwds={'groups': pd.factorize(groups)[0]})

def rr_text(fit, term='Poverty_Rate_Percent'):
    """Compute and format Rate ratio per +10 points of poverty with a 95% interval."""
    lo, hi = fit.conf_int().loc[term]
    return f"{np.exp(10 * fit.params[term]):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})"

zip_m = zip_table(HEADLINE_YEARS, 'ZIP_LOCAL')
tract_m = tract_table(HEADLINE_YEARS, 'local')
print(f"ZIP analysis: {len(zip_m)} ZIPs, {zip_m['Ped_Killed'].sum():.0f} deaths")
print(f"Tract analysis: {len(tract_m)} tracts, {tract_m['Ped_Killed'].sum():.0f} deaths, "
      f"{tract_m['ZCTA_cluster'].nunique()} ZIP clusters, {(tract_m['Ped_Killed'] == 0).mean():.0%} of tracts with no deaths")
````

````python
# Table 4, poverty quartiles at tract level
tq = tract_m.copy()
tq['PovQ'], cuts = pd.qcut(tq['Poverty_Rate_Percent'], 4, labels=['Q1', 'Q2', 'Q3', 'Q4'], retbins=True)
for q in ['Q2', 'Q3', 'Q4']:
    tq[q] = (tq['PovQ'] == q).astype(int)
print("Quartile cut points (poverty %):", [round(c, 1) for c in cuts])

def quartile_rr(outcome, controls):
    rhs = ' + '.join(['Q2', 'Q3', 'Q4'] + controls)
    fit = smf.negativebinomial(f'{outcome} ~ {rhs}', data=tq, exposure=tq['Total_Population'] * tq['n_years']).fit(
        disp=0, maxiter=500, cov_type='cluster', cov_kwds={'groups': pd.factorize(tq['ZCTA_cluster'])[0]})
    out = {}
    for q in ['Q2', 'Q3', 'Q4']:
        lo, hi = fit.conf_int().loc[q]
        out[q] = f"{np.exp(fit.params[q]):.2f} ({np.exp(lo):.2f} to {np.exp(hi):.2f})"
    return out

rows = {}
for out, label in OUTCOMES:
    rows[(label, 'no controls')] = quartile_rr(out, [])
    rows[(label, 'with controls')] = quartile_rr(out, CONTROLS)
table4 = pd.DataFrame(rows).T
print("Table 4. Rate ratio against the lowest poverty quartile (Q1)")
print(table4)
table4.to_csv(f'{OUT_DIR}/table4_tract_quartiles.csv')
````

````python
# Table 5, did the poverty gradient change between the early and late periods?
def fit_trend(panel, outcome, controls):
    panel = panel.copy()
    panel['Pov_x_Late'] = panel['Poverty_Rate_Percent'] * panel['Late']
    rhs = ' + '.join(['Poverty_Rate_Percent', 'Late', 'Pov_x_Late'] + controls)
    fit = smf.negativebinomial(f'{outcome} ~ {rhs}', data=panel,
                               exposure=panel['Total_Population'] * panel['n_years']).fit(
        disp=0, maxiter=500, cov_type='cluster', cov_kwds={'groups': pd.factorize(panel['ZCTA_cluster'])[0]})
    out = {}
    for name, expr in [('early RR', 'Poverty_Rate_Percent = 0'),
                       ('late RR', 'Poverty_Rate_Percent + Pov_x_Late = 0'),
                       ('late / early', 'Pov_x_Late = 0')]:
        t = fit.t_test(expr)
        lo, hi = np.squeeze(t.conf_int())
        eff = float(np.squeeze(t.effect))
        out[name] = f"{np.exp(10 * eff):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})"
        if name == 'late / early':
            out['p (change)'] = round(float(np.squeeze(t.pvalue)), 3)
    return out

def make_panel(early_years, late_years):
    return pd.concat([tract_table(early_years).assign(Late=0),
                      tract_table(late_years).assign(Late=1)], ignore_index=True)

panel = make_panel(EARLY_YEARS, LATE_YEARS)
rows = {}
for out, label in OUTCOMES:
    rows[(label, 'no controls')] = fit_trend(panel, out, [])
    rows[(label, 'with controls')] = fit_trend(panel, out, CONTROLS)
table5 = pd.DataFrame(rows).T
print("Table 5. Poverty rate ratio, 2019-21 against 2022-24. A late / early ratio below 1 means the gradient narrowed.")
print(table5)
table5.to_csv(f'{OUT_DIR}/table5_trend.csv')
````

````python
# Table 6, trend robustness without the pandemic years
splits = {'baseline: 2019-21 vs 2022-24': (EARLY_YEARS, LATE_YEARS),
          'drop 2020 only: 2019+2021 vs 2022-24': ([2019, 2021], [2022, 2023, 2024]),
          'drop pandemic years: 2019 vs 2023-24': ([2019], [2023, 2024])}
rows = {}
for name, (e, l) in splits.items():
    p = make_panel(e, l)
    for out, label in [('Ped_Injured', 'Injuries'), ('Ped_Killed', 'Deaths')]:
        rows[(name, label)] = fit_trend(p, out, CONTROLS)
table6 = pd.DataFrame(rows).T
print("Table 6. Trend robustness, full controls")
print(table6)
table6.to_csv(f'{OUT_DIR}/table6_trend_robustness.csv')
````

````python
# Figure 2, poverty gradient year by year (no controls)
def year_rr(year, outcome):
    m = tract_table([year])
    fit = fit_nb(m, outcome, [], m['ZCTA_cluster'])
    lo, hi = fit.conf_int().loc['Poverty_Rate_Percent']
    return np.exp(10 * fit.params['Poverty_Rate_Percent']), np.exp(10 * lo), np.exp(10 * hi)

years = list(range(STUDY_START, STUDY_END + 1))
fig, axes = plt.subplots(1, 2, figsize=(13, 4.5), sharex=True)
for ax, (out, label) in zip(axes, [('Ped_Injured', 'Pedestrian injuries'), ('Ped_Killed', 'Pedestrian deaths')]):
    est = []
    for y in years:
        try:
            est.append(year_rr(y, out))
        except Exception as e:
            print(f"{label} {y}: model failed ({type(e).__name__})")
            est.append((np.nan, np.nan, np.nan))
    est = np.array(est)
    ax.errorbar(years, est[:, 0], yerr=[est[:, 0] - est[:, 1], est[:, 2] - est[:, 0]], fmt='o-', capsize=4)
    ax.axhline(1, color='gray', linestyle='--', linewidth=1)
    ax.set_title(label)
    ax.set_xlabel('Year')
    ax.set_ylabel('Rate ratio per +10 points of poverty')
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/figure2_gradient_by_year.png', dpi=200)
plt.show()
print(f"Saved tables and figures to {OUT_DIR}")
````

### Methods

> * <font color="blue">I used NYC crash records for 2019 through 2024. The headline analysis covers 2019 through 2023 to match the five year American Community Survey estimates, and the trend analysis extends through 2024. I assigned every crash with usable coordinates to a 2020 census tract using Census boundary files, matching points that fell inside a tract or within 1,640 feet of one. This assigned about 90 percent of pedestrian deaths. I left out crashes flagged as highway incidents by road name, because a tract assigned to a highway crash doesn't really indicate where anyone lives. Tract poverty rates, populations, and the shares of workers who walk or take public transit come from ACS tables S1701 and S0801. The commercial share of floor area comes from NYC PLUTO. I kept tracts with at least 1,000 residents, which gave 2,189 tracts and 445 pedestrian deaths.</font>

> * <font color="blue">For pedestrian deaths, pedestrian injuries, and all crashes, I fit negative binomial models with population as the exposure and standard errors clustered by ZIP, because injury and crash counts vary far more than a Poisson model would allow. I report the rate ratio for each 10 point increase in a tract's poverty rate with poverty alone, with commuting shares added, and with commercial share added. I repeated the analysis at ZIP level, where 42 percent of fatal pedestrian crashes lacked a recorded ZIP. I recovered those ZIPs from coordinates and tested three ways of handling them. I checked the tract results against the highway rule, population floors from 500 to 3,000, and poverty quartiles. To look at change over time, I compared 2019 to 2021 with 2022 to 2024 using a poverty by period interaction, and I repeated the comparison without the pandemic years.</font>

### Key Insights

> * <font color="blue">Pedestrian injuries show a modest association with neighborhood poverty. Each 10 point rise in tract poverty goes with a 12 percent higher injury rate (rate ratio 1.12, 95% CI 1.08 to 1.17). After accounting for walking, transit and commercial land use, the figure falls to 5 percent (1.05, 95% CI 1.02 to 1.08) and stays between 1.04 and 1.07 across all floors and highway rules. Injury rates in the highest poverty quartile run 47 percent above the lowest, about 104 against 70 per 100k residents per year. With controls the gap narrows to 22 percent, and the second to fourth quartiles look similar to each other. The two figures bracket the answer. Which one is closer depends on whether walking and transit use are part of how poverty creates exposure or a separate confounder, and these data cannot decide between those readings.</font>

> * <font color="blue">Pedestrian deaths show no real detectable link to poverty at either scale. The tract rate ratio is 1.00 with poverty alone and 1.02 with controls. Every interval includes 1, and the quartile ratios stay between 1.03 and 1.12. Because the tract model is more precise than the ZIP model, it also rules out large effects. Deaths remain rare, with 82 percent of tracts recording none, so small effects could still be unseen here.</font>

> * <font color="blue">Overall crash rates rise with poverty only before controls. The tract rate ratio of 1.05 falls to 1.01 once commuting and land use enter the model, so I find no poverty gradient in overall crashes.</font>

> * <font color="blue">The results depend on the scale of analysis. ZIP level injury ratios of 1.14 to 1.21 shrink to 1.05 to 1.12 at tract level. ZIPs combine neighborhoods that differ sharply in commuting and land use, and tracts separate them, but I did not test that explanation, and noise in tract level commuting estimates could play a role.</font>

> * <font color="blue">The injury gradient did not narrow during the period. Comparing 2019 to 2021 with 2022 to 2024 suggests a slight narrowing (late to early ratio 0.98, 95% CI 0.96 to 1.00), but the pattern comes from the 2020 and 2021 pandemic years, when the gradient briefly rose. Dropping 2020 gives a ratio of 0.99 (p = 0.29), and comparing 2019 with 2023 to 2024 gives 1.01 (p = 0.61). Deaths show no gradient in any period.</font>

> * <font color="blue">NYC DOT's 2025 equity report found that the lowest income neighborhoods saw the largest drop in pedestrian deaths, while a 2026 analysis from the Center for an Urban Future found that safety upgrades still reach lower income neighborhoods less often and injuries remain concentrated there. My injury results fit the second account better, and my death results show no gradient for any policy to close. Without data on where treatments were installed, though, I cannot link any pattern to specific policies. An older NYC Health Department analysis from 2017 found pedestrian deaths per mile of street nearly three times as high in the poorest ZIPs. That study used 2012 to 2014 data, ZIP geography and street miles as the denominator, so differences in method could explain why my result differs.</font>

> * <font color="blue">More limitations exist; for one, a crash's tract marks where it occurred, not quite where the victim lives. Business districts still distort rates, and the analysis lacks traffic volume, road design and walking counts. About a tenth of crashes lack usable coordinates, and excluding highway crashes removes a few real local crashes. Poverty comes from a single ACS vintage for all years. The results describe association, not cause.</font>

### Recommendations

> * <font color="blue">First, agencies should direct pedestrian injury reduction funding, such as sidewalk expansion, intersection daylighting and protected crosswalks, toward high poverty neighborhoods, while sizing expectations to the modest effect I found. After accounting for commuting and land use, the gradient is closer to 5 percent per 10 points of poverty than to the raw figures. Poverty should guide where to look, and local crash data should pick the specific corridors.</font>

> * <font color="blue">Second, agencies should not use neighborhood poverty to target fatality prevention. My data show no poverty gradient in pedestrian deaths. A plausible explanation, which I did not test, is that fatal risk follows road type and traffic speed more than neighborhood income. Highway names appeared in 30 percent of fatal crashes lacking a ZIP against 5 percent of those with one, so fatality programs should draw on corridor level data about speed, road design and crash history.</font>

> * <font color="blue">Third, agencies should publish the locations and dates of street safety treatments and speed limit changes in a form that analysts can link to crash records. Without it, no one outside the agency can tell whether Vision Zero narrowed or widened neighborhood gaps, and the available reports disagree.</font>

> * <font color="blue">Fourth, agencies should standardize location entry in police crash reports. ZIP codes were missing for 34 percent of all crashes and 42 percent of fatal pedestrian crashes, and the share of pedestrian deaths without one rose from 31 percent in 2019 to between 44 and 47 percent in 2022 and 2023. Consistent coordinates and ZIPs would make local trends easier to track.</font>

> * <font color="blue">Fifth, agencies should link crash records to victim residence, for example through EMS or hospital data. My analysis shows where crashes happen, and many pedestrian harms may fall on people who live elsewhere.</font>

### <font color="blue">Key terms</font>

- <font color="blue">**Site.** A cluster of crashes within 40 feet of each other, usually one intersection.</font>
- <font color="blue">**Harm score.** Each crash counts at most one death (weighted ten times an injury) and three injuries, so one mass casualty crash cannot dominate the ranking.</font>
- <font color="blue">**Fatal crashes per 1,000.** Among crashes that harmed a pedestrian, how many involved a death.</font>
- <font color="blue">**Matched untreated sites.** Intersections without an LPI that had similar crash counts in the two windows before the treated site's installation date.</font>
- <font color="blue">**Placebo check.** The same analysis run as if each LPI had been installed two years earlier. It should give a ratio near 1.</font>
- <font color="blue">**Adjusted ratio.** The actual ratio divided by the placebo ratio. It assumes the gap between treated and untreated sites was the same in both periods, so I treat it as exploratory.</font>

### <font color="blue">Interpretation</font>

<font color="blue">The LPI table shows the actual ratio, the placebo ratio and the adjusted ratio for each group. An adjusted ratio below 1 indicates fewer crashes than expected after installation. The placebo came back well above 1, so the estimate carries that caveat.</font>

````python
# Settings and flags
from scipy.spatial import cKDTree
from sklearn.cluster import DBSCAN
from statsmodels.stats.proportion import proportion_confint

VRU_YEARS = list(range(2019, 2025))
DEATH_WEIGHT = 10                 # a death counts as this many injuries when ranking sites.
CAP_KILLED, CAP_INJURED = 1, 3    # per crash caps for ranking, so one mass casualty crash cannot dominate
CLUSTER_FT = 40                   # crashes within this many feet of each other form one site
WINDOW_YEARS = 2                  # LPI evaluation, years before and after installation
RADIUS_FT = 100                   # crashes within this distance count as at the intersection
EXCL_FT = 300                     # control sites must be farther than this from any LPI
N_CONTROLS = 20000                # random sample of control sites, for speed
HEAVY_WORDS = ['BOX TRUCK', 'TRACTOR', 'DUMP', 'FLAT BED', 'TANKER', 'CONCRETE', 'GARBAGE', 'REFUSE',
               'BUS', 'LARGE COM', 'STAKE', 'BEVERAGE', 'REFRIGERATED', 'CHASSIS']   # edit to match the labels you see

fac_cols = [f'CONTRIBUTING FACTOR VEHICLE {i}' for i in range(1, 6)]
veh_cols = [f'VEHICLE TYPE CODE {i}' for i in range(1, 6)]
NEEDED = ['NUMBER OF PEDESTRIANS KILLED', 'NUMBER OF PEDESTRIANS INJURED',
          'NUMBER OF CYCLIST KILLED', 'NUMBER OF CYCLIST INJURED',
          'ON STREET NAME', 'CROSS STREET NAME', 'CRASH TIME'] + fac_cols + veh_cols
missing = [c for c in NEEDED if c not in crashes.columns]
if missing:
    raise KeyError(f"These columns are not in your crash table: {missing}")

for c in NEEDED[:4]:
    crashes[c] = crashes[c].fillna(0)
crashes['PED_killed'] = crashes['NUMBER OF PEDESTRIANS KILLED']
crashes['PED_injured'] = crashes['NUMBER OF PEDESTRIANS INJURED']
crashes['CYC_killed'] = crashes['NUMBER OF CYCLIST KILLED']
crashes['CYC_injured'] = crashes['NUMBER OF CYCLIST INJURED']
crashes['PED_harm'] = (crashes['PED_killed'] + crashes['PED_injured']) > 0
crashes['CYC_harm'] = (crashes['CYC_killed'] + crashes['CYC_injured']) > 0
for grp in ['PED', 'CYC']:
    crashes[f'{grp}_capped'] = (DEATH_WEIGHT * np.minimum(crashes[f'{grp}_killed'], CAP_KILLED)
                                + np.minimum(crashes[f'{grp}_injured'], CAP_INJURED))

crashes['X'] = pts.geometry.x.reindex(crashes.index)         # projected coordinates in feet, from Milestone 5
crashes['Y'] = pts.geometry.y.reindex(crashes.index)
on = crashes['ON STREET NAME'].fillna('').str.upper().str.strip()
cross = crashes['CROSS STREET NAME'].fillna('').str.upper().str.strip()
crashes['site_key'] = [' & '.join(sorted((a, b))) if a and b else None for a, b in zip(on, cross)]   # label only

inwin = crashes['YEAR'].isin(VRU_YEARS)
print(f"Pedestrian harm crashes {VRU_YEARS[0]} to {VRU_YEARS[-1]}: {(inwin & crashes['PED_harm']).sum():,} "
      f"({crashes.loc[inwin, 'PED_killed'].sum():.0f} deaths)")
print(f"Cyclist harm crashes: {(inwin & crashes['CYC_harm']).sum():,} ({crashes.loc[inwin, 'CYC_killed'].sum():.0f} deaths)")
````

````python
# Sites ranked by harm, separately for pedestrians and cyclists
def mode_or_nan(s):
    s = s.dropna()
    return s.mode().iat[0] if len(s) else np.nan

def build_sites(grp):
    flag, capped_col = f'{grp}_harm', f'{grp}_capped'
    mask = crashes['YEAR'].isin(VRU_YEARS) & crashes[flag]
    d = crashes[mask & crashes['X'].notna()].copy()
    d['X'], d['Y'] = d['X'].round(0), d['Y'].round(0)
    uniq = d[['X', 'Y']].drop_duplicates().reset_index(drop=True)
    uniq['cluster'] = DBSCAN(eps=CLUSTER_FT, min_samples=1).fit_predict(uniq[['X', 'Y']].values)
    d = d.merge(uniq, on=['X', 'Y'], how='left')
    s = d.groupby('cluster').agg(
        crashes=('cluster', 'size'), killed=(f'{grp}_killed', 'sum'), injured=(f'{grp}_injured', 'sum'),
        harm=(capped_col, 'sum'), hwy_share=('is_highway', 'mean'), x=('X', 'median'), y=('Y', 'median'),
        tract=('TRACT', 'first'), site=('site_key', mode_or_nan), on_street=('ON STREET NAME', mode_or_nan))
    s['label'] = s['site'].fillna(s['on_street'])
    s = s.sort_values('harm', ascending=False).reset_index(drop=True)
    return s, d[capped_col].sum() / crashes.loc[mask, capped_col].sum()

sites_ped, cover_ped = build_sites('PED')
sites_cyc, cover_cyc = build_sites('CYC')
poverty = tract_acs.set_index('GEOID')['Poverty_Rate_Percent']

for name, s, cover in [('Pedestrian', sites_ped, cover_ped), ('Cyclist', sites_cyc, cover_cyc)]:
    total = s['harm'].sum()
    s['poverty'] = s['tract'].map(poverty)
    print(f"\n{name} harm: {len(s):,} sites, {cover:.1%} of capped harm has usable coordinates")
    for pct in [0.01, 0.05, 0.10]:
        k = max(1, int(len(s) * pct))
        print(f"  Worst {pct:.0%} of sites ({k:,}) hold {s['harm'].head(k).sum() / total:.1%} of harm")
    print(f"  Median tract poverty (%), worst 100 sites {s['poverty'].head(100).median():.1f}, "
          f"all sites {s['poverty'].median():.1f}, all NYC tracts {poverty.median():.1f}")
    print(s[['label', 'crashes', 'killed', 'injured', 'harm', 'hwy_share']].head(15).round(2).to_string(index=False))
    s.head(500).to_csv(f"{OUT_DIR}/m6_sites_{name.lower()}.csv", index=False)
print("\nhwy_share comes from a street name keyword and wrongly flags some boulevards, such as Eastern Parkway.")
````

````python
# How deadly is each vehicle type and factor for pedestrians? Fatal crashes per 1,000 harmed pedestrian crashes
ped = crashes[crashes['YEAR'].isin(VRU_YEARS) & crashes['PED_harm']].copy()
ped['fatal'] = ped['PED_killed'] > 0
veh_text = ped[veh_cols].apply(lambda s: s.astype(str).str.upper())
heavy_re = '|'.join(HEAVY_WORDS)
ped['heavy_any'] = veh_text.apply(lambda s: s.str.contains(heavy_re, regex=True) & ~s.str.contains('PICK')).any(axis=1)

def severity_by_slot(frame, cols, drop, min_n):
    long = (frame[['fatal'] + cols].rename_axis('row').reset_index()
            .melt(id_vars=['row', 'fatal'], value_vars=cols, value_name='val').dropna(subset=['val']))
    long['val'] = long['val'].astype(str).str.strip()
    long = long[~long['val'].str.upper().isin(drop)].drop_duplicates(['row', 'val'])
    g = long.groupby('val').agg(n=('row', 'size'), fatal=('fatal', 'sum'))
    g = g[g['n'] >= min_n].copy()
    lo, hi = proportion_confint(g['fatal'], g['n'], method='wilson')
    g['per_1000'], g['lo'], g['hi'] = g['fatal'] / g['n'] * 1000, lo * 1000, hi * 1000
    return g.sort_values('n', ascending=False)

print(f"Pedestrian harm crashes: {len(ped):,}, fatal: {ped['fatal'].sum():,} ({ped['fatal'].mean() * 1000:.1f} per 1,000)")
for flag, label in [(True, 'Heavy vehicle involved'), (False, 'No heavy vehicle')]:
    sub = ped[ped['heavy_any'] == flag]
    lo, hi = proportion_confint(sub['fatal'].sum(), len(sub), method='wilson')
    print(f"  {label}: {len(sub):,} crashes, {sub['fatal'].mean() * 1000:.1f} fatal per 1,000 "
          f"(95% CI {lo * 1000:.1f} to {hi * 1000:.1f})")

veh_sev = severity_by_slot(ped, veh_cols, drop=('UNKNOWN', 'UNSPECIFIED'), min_n=300)
fac_sev = severity_by_slot(ped, fac_cols, drop=('UNSPECIFIED',), min_n=300)
print("\nBy vehicle type (at least 300 crashes)")
print(veh_sev.head(15).round(1))
print("\nBy contributing factor (at least 300 crashes)")
print(fac_sev.head(15).round(1))
veh_sev.to_csv(f'{OUT_DIR}/m6_severity_vehicle.csv')
fac_sev.to_csv(f'{OUT_DIR}/m6_severity_factor.csv')
````

````python
# LPI evaluation, data preparation
r = requests.get("https://data.cityofnewyork.us/resource/xc4v-ntf4.csv", params={'$limit': 50000}, timeout=180)
r.raise_for_status()
lpi = pd.read_csv(io.StringIO(r.text))
lpi['T'] = pd.to_datetime(lpi['install_da'], errors='coerce')
lpi = lpi.dropna(subset=['T', 'lat', 'long']).copy()
lg = gpd.GeoDataFrame(lpi, geometry=gpd.points_from_xy(lpi['long'], lpi['lat']), crs=4326).to_crs(2263)
lpi['x'], lpi['y'] = lg.geometry.x.values, lg.geometry.y.values
lpi = lpi.reset_index(drop=True)
lpi['grp'] = DBSCAN(eps=60, min_samples=1).fit_predict(lpi[['x', 'y']].values)
lpi = (lpi.sort_values('T').groupby('grp')
          .agg(T=('T', 'min'), x=('x', 'median'), y=('y', 'median'), n=('T', 'size')).reset_index())   # earliest install per intersection
lpi_days = ((lpi['T'] - pd.Timestamp('1970-01-01')).dt.days).values
lpi_xy = lpi[['x', 'y']].values
print(f"LPI intersections: {len(lpi):,}, installed {lpi['T'].min().date()} to {lpi['T'].max().date()}")

# Pedestrian harm crashes across every year in the file, projected to feet
pc = data.copy()
pc['CRASH DATE'] = pd.to_datetime(pc['CRASH DATE'])
pk = pc['NUMBER OF PEDESTRIANS KILLED'].fillna(0)
pi = pc['NUMBER OF PEDESTRIANS INJURED'].fillna(0)
ok = ((pk + pi) > 0) & pc['LATITUDE'].between(40.4, 41.0) & pc['LONGITUDE'].between(-74.3, -73.6)
pc = pc[ok].copy()
cg = gpd.GeoDataFrame(geometry=gpd.points_from_xy(pc['LONGITUDE'], pc['LATITUDE']), crs=4326).to_crs(2263)
xy = np.c_[cg.geometry.x.values, cg.geometry.y.values]
tdays = ((pc['CRASH DATE'] - pd.Timestamp('1970-01-01')).dt.days).values
hour = pd.to_datetime(pc['CRASH TIME'], format='%H:%M', errors='coerce').dt.hour.values
DATA_MIN, DATA_MAX = tdays.min(), tdays.max()
print(f"Pedestrian harm crashes with coordinates: {len(pc):,}, "
      f"{pc['CRASH DATE'].min().date()} to {pc['CRASH DATE'].max().date()}")

# Candidate control sites, meaning places where a pedestrian was harmed, away from every LPI
uq = pd.DataFrame(np.round(xy), columns=['x', 'y']).drop_duplicates().reset_index(drop=True)
uq['cl'] = DBSCAN(eps=40, min_samples=1).fit_predict(uq[['x', 'y']].values)
cpts = uq.groupby('cl')[['x', 'y']].median().values
cpts = cpts[cKDTree(lpi_xy).query(cpts)[0] > EXCL_FT]
cpts = cpts[np.random.default_rng(11).choice(len(cpts), size=min(N_CONTROLS, len(cpts)), replace=False)]
print(f"Control sites used: {len(cpts):,}")

masks = {'all hours': np.ones(len(xy), bool),
         'daytime 7am to 7pm': (hour >= 7) & (hour < 19),
         'night': (hour < 7) | (hour >= 19)}
nbrs = {}
for name, m in masks.items():
    tree, d = cKDTree(xy[m]), tdays[m]
    nbrs[name] = tuple([np.sort(d[np.asarray(i, dtype=int)]) for i in tree.query_ball_point(p, r=RADIUS_FT)]
                       for p in (lpi_xy, cpts))
````

````python
# LPI evaluation, matched on two pre-installation windows, with a placebo check
# For each LPI intersection, count pedestrian harm crashes within RADIUS_FT in the WINDOW_YEARS after installation.
# Compare with untreated sites matched on installation year and on the crash counts in the two preceding windows.
# The placebo pretends each LPI arrived WINDOW_YEARS earlier and should give a ratio near 1.

BIN_EDGES = [1, 2]          # each pre window count falls in a stratum of 0, 1, or 2 or more
ctrl_cache = {}

def count_in(arrs, lo, hi):
    return np.array([np.searchsorted(a, h) - np.searchsorted(a, l) for a, l, h in zip(arrs, lo, hi)])

def ctrl_counts(mask_name, t0, w):
    key = (mask_name, t0)
    if key not in ctrl_cache:
        c_arrs = nbrs[mask_name][1]
        lo = np.full(len(c_arrs), t0)
        ctrl_cache[key] = (count_in(c_arrs, lo - 2 * w, lo - w), count_in(c_arrs, lo - w, lo), count_in(c_arrs, lo, lo + w))
    return ctrl_cache[key]

def unit_table(mask_name, shift_years, treated_filter=None, min_controls=20):
    t_arrs = nbrs[mask_name][0]
    w, sh = int(WINDOW_YEARS * 365.25), int(shift_years * 365.25)
    T = lpi_days - sh
    ok = (T - 2 * w >= DATA_MIN) & (T + w <= DATA_MAX)
    if treated_filter is not None:
        ok &= treated_filter
    ti = np.where(ok)[0]
    arr, Tt = [t_arrs[i] for i in ti], T[ti]
    tdf = pd.DataFrame({'unit': ti, 'cohort': pd.to_datetime(Tt, unit='D').year.values,
                        'a': count_in(arr, Tt - 2 * w, Tt - w), 'b': count_in(arr, Tt - w, Tt),
                        'post': count_in(arr, Tt, Tt + w)})
    tdf = tdf[(tdf['a'] + tdf['b']) >= 1].copy()
    tdf['ba'], tdf['bb'] = np.digitize(tdf['a'], BIN_EDGES), np.digitize(tdf['b'], BIN_EDGES)
    rows = []
    for y in tdf['cohort'].unique():
        t0 = int((pd.Timestamp(int(y), 7, 1) - pd.Timestamp('1970-01-01')).days)
        if t0 - 2 * w < DATA_MIN or t0 + w > DATA_MAX:
            continue
        ca, cb, cp = ctrl_counts(mask_name, t0, w)
        m = (ca + cb) >= 1
        rows.append(pd.DataFrame({'cohort': y, 'ba': np.digitize(ca[m], BIN_EDGES), 'bb': np.digitize(cb[m], BIN_EDGES),
                                  'a': ca[m], 'b': cb[m], 'post': cp[m]}))
    ctrl = pd.concat(rows)
    cs = (ctrl.groupby(['cohort', 'ba', 'bb'])
              .agg(n=('post', 'size'), exp=('post', 'mean'), exp_a=('a', 'mean'), exp_b=('b', 'mean')).reset_index())
    tdf = tdf.merge(cs, on=['cohort', 'ba', 'bb'], how='left')
    return tdf[tdf['n'] >= min_controls].rename(columns={'post': 'obs'})

def summarize(actual, placebo, n_boot=1000, seed=7):
    common = np.intersect1d(actual['unit'], placebo['unit'])
    A, P = actual.set_index('unit').loc[common], placebo.set_index('unit').loc[common]
    n = len(common)
    bs = np.random.default_rng(seed).integers(0, n, size=(n_boot, n))
    ra = A['obs'].values[bs].sum(axis=1) / A['exp'].values[bs].sum(axis=1)
    rp = P['obs'].values[bs].sum(axis=1) / P['exp'].values[bs].sum(axis=1)
    out = {'intersections': n, 'observed post': int(A['obs'].sum()), 'expected post': round(A['exp'].sum(), 1),
           'RR actual': A['obs'].sum() / A['exp'].sum(), 'RR placebo': P['obs'].sum() / P['exp'].sum()}
    out['RR adjusted'] = out['RR actual'] / out['RR placebo']
    for key, boot in [('actual', ra), ('placebo', rp), ('adjusted', ra / rp)]:
        out[f'{key} lo'], out[f'{key} hi'] = np.percentile(boot, [2.5, 97.5])
    out['pre balance'] = round((A['a'] + A['b']).sum() / (A['exp_a'] + A['exp_b']).sum(), 3)
    return out

def show(results):
    df = pd.DataFrame(results).T
    for key, col in [('actual', 'RR actual'), ('placebo', 'RR placebo'), ('adjusted', 'RR adjusted')]:
        df[f'{col} (95% CI)'] = [f"{v:.2f} ({lo:.2f} to {hi:.2f})"
                                 for v, lo, hi in zip(df[col], df[f'{key} lo'], df[f'{key} hi'])]
    return df

COLS = ['intersections', 'observed post', 'expected post', 'RR actual (95% CI)',
        'RR placebo (95% CI)', 'RR adjusted (95% CI)', 'pre balance']
res = {name: summarize(unit_table(name, 0), unit_table(name, WINDOW_YEARS)) for name in nbrs}
table = show(res)
print(f"Pedestrian harm crashes within {RADIUS_FT} ft, {WINDOW_YEARS} years after installation against matched untreated sites")
print("The placebo should sit near 1. The adjusted ratio divides the actual ratio by the placebo ratio.")
print(table[COLS])
table.to_csv(f'{OUT_DIR}/m6_lpi.csv')
````

````python
# Tract by year panel
j = gpd.sjoin_nearest(cg, tracts, how='left', max_distance=SNAP_FT)
j = j[~j.index.duplicated(keep='first')]
pcs = pc.reset_index(drop=True)
pcs['GEOID'] = j['GEOID'].reindex(pcs.index)
pcs['YEAR'] = pcs['CRASH DATE'].dt.year
pcs['killed'] = pcs['NUMBER OF PEDESTRIANS KILLED'].fillna(0)
cnt = (pcs.dropna(subset=['GEOID']).groupby(['GEOID', 'YEAR'])
          .agg(harm=('YEAR', 'size'), killed=('killed', 'sum')).reset_index())

area = tracts.assign(area_sqmi=tracts.geometry.area / 27_878_400)[['GEOID', 'area_sqmi']]
static = (tract_acs[['GEOID', 'Total_Population', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share']]
          .merge(pluto_tract, on='GEOID', how='left').merge(tract_cluster, on='GEOID', how='left')
          .merge(area, on='GEOID', how='left'))
static = static[static['Total_Population'] >= MIN_POP_TRACT].dropna(
    subset=['Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share', 'Commercial_Share', 'ZCTA_cluster', 'area_sqmi']).copy()
static['density'] = static['Total_Population'] / static['area_sqmi']
county = static['GEOID'].str[2:5]
for code, nm in {'061': 'boro_manhattan', '005': 'boro_bronx', '047': 'boro_brooklyn', '081': 'boro_queens'}.items():
    static[nm] = (county == code).astype(int)                # Staten Island is the reference
QUARTILES = ['Q1 lowest', 'Q2', 'Q3', 'Q4 highest']
static['pov_quartile'] = pd.qcut(static['Poverty_Rate_Percent'], 4, labels=QUARTILES)

idx = pd.MultiIndex.from_product([static['GEOID'].values, YEARS_ALL], names=['GEOID', 'YEAR'])
g = cnt.set_index(['GEOID', 'YEAR']).reindex(idx, fill_value=0).reset_index()
harm_w = g.pivot(index='GEOID', columns='YEAR', values='harm')
kill_w = g.pivot(index='GEOID', columns='YEAR', values='killed')

def lag(w, Y, k):
    return w[Y - k].values if (Y - k) in w.columns else np.full(len(w), np.nan)

rows = []
for Y in TARGET_YEARS:
    f = pd.DataFrame({'GEOID': harm_w.index, 'YEAR': Y, 'y': harm_w[Y].values, 'deaths': kill_w[Y].values,
                      'lag1': lag(harm_w, Y, 1), 'lag2': lag(harm_w, Y, 2), 'lag3': lag(harm_w, Y, 3),
                      'lag4': lag(harm_w, Y, 4), 'lag5': lag(harm_w, Y, 5),     # lags 4 and 5 feed the baselines only
                      'kill3': kill_w[[Y - 1, Y - 2, Y - 3]].sum(axis=1).values})
    f['avg3'] = (f['lag1'] + f['lag2'] + f['lag3']) / 3
    rows.append(f)
panel = pd.concat(rows).merge(static, on='GEOID', how='inner').reset_index(drop=True)

HISTORY = ['lag1', 'lag2', 'lag3', 'avg3', 'kill3']
NEIGHBORHOOD = ['Total_Population', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share', 'Commercial_Share',
                'density', 'boro_manhattan', 'boro_bronx', 'boro_brooklyn', 'boro_queens']
print(f"Panel rows: {len(panel):,}, tracts: {panel['GEOID'].nunique():,}, ZIP clusters: {panel['ZCTA_cluster'].nunique()}")
print(panel.groupby('YEAR')[['y', 'deaths']].sum().T.to_string())
print(f"Share of tract-years with no pedestrian harm: {(panel['y'] == 0).mean():.0%}")
````

````python
# Models and baselines, out of time and out of space
MODELS = [('GBM history only', HISTORY, 'gbm'),
          ('GBM neighborhood only', NEIGHBORHOOD, 'gbm'),
          ('GBM history + neighborhood', HISTORY + NEIGHBORHOOD, 'gbm'),
          ('Poisson regression history + neighborhood', HISTORY + NEIGHBORHOOD, 'glm')]

def make_model(kind):
    if kind == 'gbm':
        return HistGradientBoostingRegressor(loss='poisson', max_iter=250, learning_rate=0.05, max_depth=4,
                                             min_samples_leaf=50, l2_regularization=1.0, random_state=0)
    return make_pipeline(StandardScaler(), PoissonRegressor(alpha=1e-3, max_iter=500))

def fit_predict(train, test, feats, kind):
    w = train['Total_Population'].values
    model = make_model(kind)                              # Poisson with exposure, written as a weighted rate model
    if kind == 'gbm':
        model.fit(train[feats], train['y'].values / w, sample_weight=w)
    else:
        model.fit(train[feats], train['y'].values / w, poissonregressor__sample_weight=w)
    return np.clip(model.predict(test[feats]), 1e-9, None) * test['Total_Population'].values

clusters = np.random.default_rng(42).permutation(panel['ZCTA_cluster'].unique())
parts = []
for held in np.array_split(clusters, 5):
    is_held = panel['ZCTA_cluster'].isin(held)
    train = panel[~is_held & (panel['YEAR'] <= TRAIN_END)]
    test = panel[is_held & panel['YEAR'].isin(TEST_YEARS)].copy()
    for name, feats, kind in MODELS:
        test[name] = fit_predict(train, test, feats, kind)
    parts.append(test)
test = pd.concat(parts)

LAGS = [f'lag{k}' for k in range(1, 6)]
assert test[LAGS].notna().all().all(), "Baselines need five lag years in every test row"
wts = DECAY ** np.arange(1, 6)
test['Last year'] = test['lag1']
test['Three year average'] = test['avg3']
test['Five year average'] = test[LAGS].mean(axis=1)
test['Recency weighted'] = (test[LAGS].values * wts).sum(axis=1) / wts.sum()
BASELINES = ['Last year', 'Three year average', 'Five year average', 'Recency weighted']
METHODS = BASELINES + [m[0] for m in MODELS]
print(f"Test rows: {len(test):,}, test years {TEST_YEARS}, tracts never seen in training")

# Arrays of shape (tracts, test years) make the bootstrap fast
tr_ids = np.sort(test['GEOID'].unique())
arr = {c: test.pivot(index='GEOID', columns='YEAR', values=c).loc[tr_ids].values for c in METHODS + ['y', 'deaths']}
n, ny = arr['y'].shape

def topmask(score, k):
    m = np.zeros(score.shape, bool)
    for jy in range(score.shape[1]):
        m[np.argsort(-score[:, jy])[:k], jy] = True
    return m

def capture(score, target, top, idx=None, jit=None):
    """Average over years of the share of the target that falls in the top fraction of tracts ranked by score."""
    if idx is not None:
        score, target = score[idx], target[idx]
    if jit is not None:
        score = score + jit                                   # random tie breaking
    k = int(round(len(score) * top))
    return np.mean([target[np.argsort(-score[:, jy])[:k], jy].sum() / target[:, jy].sum() for jy in range(score.shape[1])])

jit0 = np.random.default_rng(0).random((n, ny)) * 1e-6
res = {}
for m in METHODS + ['Best possible (oracle)']:
    col = 'y' if m.startswith('Best') else m
    row = {'harm in top 5%': capture(arr[col], arr['y'], 0.05, jit=jit0),
           'harm in top 10%': capture(arr[col], arr['y'], 0.10, jit=jit0),
           'harm in top 20%': capture(arr[col], arr['y'], 0.20, jit=jit0),
           'deaths in top 10%': capture(arr[col], arr['deaths'], 0.10, jit=jit0),
           'MAE of counts': mean_absolute_error(test['y'], test[col])}
    if m in [x[0] for x in MODELS]:
        row['Poisson deviance'] = mean_poisson_deviance(test['y'], np.clip(test[m], 1e-6, None))
    res[m] = row
table = pd.DataFrame(res).T
table.loc['Random ranking', ['harm in top 5%', 'harm in top 10%', 'harm in top 20%', 'deaths in top 10%']] = [0.05, 0.10, 0.20, 0.10]
pd.set_option('display.width', 250, 'display.max_columns', 30, 'display.max_colwidth', 45)
print(table.round(3))
table.to_csv(f'{OUT_DIR}/m7_model_comparison.csv')
test.drop(columns=[c for c in test.columns if c.startswith('boro_')]).to_csv(f'{OUT_DIR}/m7_test_predictions.csv', index=False)
````

````python
# Bootstrap intervals for the top 10 percent capture and for differences from the three year average
rng = np.random.default_rng(7)
boot = {m: np.empty(N_BOOT) for m in METHODS}
for b in range(N_BOOT):
    ib = rng.integers(0, n, n)
    jb = rng.random((n, ny)) * 1e-6
    for m in METHODS:
        boot[m][b] = capture(arr[m], arr['y'], TOP, ib, jb)

REF = 'Three year average'
rows = []
for m in METHODS:
    point = capture(arr[m], arr['y'], TOP, jit=jit0)
    lo, hi = np.percentile(boot[m], [2.5, 97.5])
    d = boot[m] - boot[REF]
    dlo, dhi = np.percentile(d, [2.5, 97.5])
    rows.append({'method': m, 'harm in top 10%': f"{point:.3f} ({lo:.3f} to {hi:.3f})",
                 f'difference from {REF.lower()}': '' if m == REF else f"{point - capture(arr[REF], arr['y'], TOP, jit=jit0):+.3f} ({dlo:+.3f} to {dhi:+.3f})",
                 'share of resamples that beat it': '' if m == REF else f"{(d > 0).mean():.0%}"})
boot_table = pd.DataFrame(rows).set_index('method')
print("Share of next-year pedestrian harm in the top 10% of tracts, with 95% bootstrap intervals")
print(boot_table)
boot_table.to_csv(f'{OUT_DIR}/m7_bootstrap_capture.csv')
````

````python
# Figures, gain curves and interval plot
STYLE = {'Three year average': dict(color=PAL[0], ls='-', marker='o'),
         'GBM history + neighborhood': dict(color=PAL[1], ls='--', marker='s'),
         'GBM neighborhood only': dict(color=PAL[2], ls='-.', marker='^')}
fracs = np.linspace(0.01, 0.5, 25)
fig, ax = plt.subplots(figsize=(7.5, 5.5))
for m, st in STYLE.items():
    ax.plot(fracs * 100, [capture(arr[m], arr['y'], f, jit=jit0) * 100 for f in fracs], markevery=4, label=m, **st)
ax.plot(fracs * 100, [capture(arr['y'], arr['y'], f, jit=jit0) * 100 for f in fracs], color='gray', ls=':', label='Best possible')
ax.plot(fracs * 100, fracs * 100, color='black', ls='--', lw=1, label='Random')
ax.set_xlabel('Share of tracts flagged, highest predicted first (%)')
ax.set_ylabel('Share of next-year pedestrian harm captured (%)')
ax.set_title('Capturing next-year pedestrian harm')
ax.legend()
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/m7_gain_curves.png', dpi=200)
plt.show()
print("Figure description. Gain curves for tracts the models never saw. The three year average and the combined model "
      "overlap and sit just below the best possible line. The neighborhood only model sits lower but above random.")

order = METHODS[::-1]
fig, ax = plt.subplots(figsize=(8, 5))
for k, m in enumerate(order):
    point = capture(arr[m], arr['y'], TOP, jit=jit0)
    lo, hi = np.percentile(boot[m], [2.5, 97.5])
    ax.errorbar(point, k, xerr=[[point - lo], [hi - point]], fmt='o' if m in BASELINES else 's', capsize=4, color=PAL[0] if m in BASELINES else PAL[1])
ax.axvline(0.10, color='black', ls='--', lw=1)
ax.axvline(capture(arr['y'], arr['y'], TOP, jit=jit0), color='gray', ls=':')
ax.set_yticks(range(len(order)))
ax.set_yticklabels(order)
ax.set_xlabel('Share of next-year harm in the top 10% of tracts (95% interval)')
ax.set_title('Baselines (circles) and models (squares)')
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/m7_capture_intervals.png', dpi=200)
plt.show()
print("Figure description. Capture rates with intervals for each method. The dashed line is random ranking and the dotted "
      "line is the best possible. Intervals for the simple averages and the combined model overlap heavily.")
````

````python
# Equity audit by poverty quartile, with bootstrap intervals
pq = static.set_index('GEOID').loc[tr_ids, 'pov_quartile'].cat.codes.values        # 0 is the lowest poverty quartile
k_top = int(round(n * TOP))
AUDIT = ['Three year average', 'GBM neighborhood only', 'GBM history + neighborhood']

def audit_stats(score, ytab, qv, jit_s, jit_a):
    act, fl = topmask(ytab + jit_a, k_top), topmask(score + jit_s, k_top)
    rec, prec = np.empty(4), np.empty(4)
    for q in range(4):
        mq = (qv[:, None] == q)
        hit = (act & fl & mq).sum()
        rec[q] = hit / max((act & mq).sum(), 1)
        prec[q] = hit / max((fl & mq).sum(), 1)
    return rec, prec

point = {m: audit_stats(arr[m], arr['y'], pq, jit0, jit0[::-1]) for m in AUDIT}
rng = np.random.default_rng(11)
brec = {m: np.empty((N_BOOT, 4)) for m in AUDIT}
bprec = {m: np.empty((N_BOOT, 4)) for m in AUDIT}
for b in range(N_BOOT):
    ib = rng.integers(0, n, n)
    ja, js = rng.random((n, ny)) * 1e-6, rng.random((n, ny)) * 1e-6
    for m in AUDIT:
        brec[m][b], bprec[m][b] = audit_stats(arr[m][ib], arr['y'][ib], pq[ib], js, ja)

def ci(a):
    lo, hi = np.percentile(a, [2.5, 97.5], axis=0)
    return lo, hi

harm_share = np.array([arr['y'][pq == q].sum() / arr['y'].sum() for q in range(4)])
audit = pd.DataFrame({'share of actual harm': harm_share}, index=QUARTILES)
for m in AUDIT:
    rl, rh = ci(brec[m])
    pl, ph = ci(bprec[m])
    audit[f'{m}: recall'] = [f"{point[m][0][q]:.2f} ({rl[q]:.2f} to {rh[q]:.2f})" for q in range(4)]
    audit[f'{m}: precision'] = [f"{point[m][1][q]:.2f} ({pl[q]:.2f} to {ph[q]:.2f})" for q in range(4)]
    audit[f'{m}: predicted / actual'] = [arr[m][pq == q].sum() / arr['y'][pq == q].sum() for q in range(4)]
print("Recall is the share of the truly worst 10% of tracts that the method flagged.")
print("Precision is the share of flagged tracts that truly belong to the worst 10%.")
print(audit.round(2).T)
audit.to_csv(f'{OUT_DIR}/m7_equity_audit.csv')

print("\nGap in recall, highest poverty quartile minus lowest (positive means the method finds poor tracts better)")
for m in AUDIT:
    gap = brec[m][:, 3] - brec[m][:, 0]
    lo, hi = np.percentile(gap, [2.5, 97.5])
    print(f"  {m}: {point[m][0][3] - point[m][0][0]:+.3f} ({lo:+.3f} to {hi:+.3f})")

fig, ax = plt.subplots(figsize=(9, 4.8))
w, hatches = 0.26, ['', '//', 'xx']
for k, m in enumerate(AUDIT):
    rl, rh = ci(brec[m])
    v = point[m][0]
    ax.bar(np.arange(4) + (k - 1) * w, v, w, yerr=[v - rl, rh - v], capsize=3, hatch=hatches[k], color=PAL[k],
           edgecolor='black', label=m)
ax.set_xticks(range(4))
ax.set_xticklabels(QUARTILES)
ax.set_ylabel('Recall of the worst 10% of tracts')
ax.set_xlabel('Tract poverty quartile')
ax.set_title('Does the method find high-harm tracts equally well across poverty levels?')
ax.legend()
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/m7_equity_recall.png', dpi=200)
plt.show()
print("Figure description. Bars show how many of the truly worst tracts each method flagged, by poverty quartile, with 95% "
      "intervals. The three year average and the combined model look similar in every quartile, and the neighborhood only "
      "model is lower in every quartile.")
````

````python
# Which features matter? Permutation importance for the combined model
train_all = panel[panel['YEAR'] <= TRAIN_END]
test_all = panel[panel['YEAR'].isin(TEST_YEARS)]
feats = HISTORY + NEIGHBORHOOD
w_tr = train_all['Total_Population'].values
final = make_model('gbm')
final.fit(train_all[feats], train_all['y'].values / w_tr, sample_weight=w_tr)
pi_res = permutation_importance(final, test_all[feats], test_all['y'].values / test_all['Total_Population'].values,
                                scoring='neg_mean_poisson_deviance', sample_weight=test_all['Total_Population'].values,
                                n_repeats=5, random_state=0)
imp = (pd.DataFrame({'feature': feats, 'importance': pi_res.importances_mean, 'sd': pi_res.importances_std})
         .sort_values('importance', ascending=False))
print("Increase in Poisson deviance when each feature is shuffled. Larger means more important.")
print(imp.round(5).to_string(index=False))
imp.to_csv(f'{OUT_DIR}/m7_permutation_importance.csv', index=False)

top8 = imp.head(8).iloc[::-1]
fig, ax = plt.subplots(figsize=(7, 4.2))
ax.barh(top8['feature'], top8['importance'], xerr=top8['sd'], color=PAL[0], edgecolor='black')
ax.set_xlabel('Increase in Poisson deviance when shuffled')
ax.set_title('Most important features')
plt.tight_layout()
plt.savefig(f'{OUT_DIR}/m7_importance.png', dpi=200)
plt.show()
print("Figure description. Horizontal bars rank features by how much shuffling them hurts the model. The three year average "
      "and tract population matter most, and the poverty rate matters little.")
print(f"Saved tables and figures to {OUT_DIR}")
````

### <font color='blue'>Research question and why</font>

<font color='blue'>Does a model that uses neighborhood characteristics predict next-year pedestrian harm better than crash history alone, and is its accuracy equal across poverty levels?</font>

<font color='blue'>Planners often decide where to act by ranking neighborhoods on recent crash counts. I wanted to know whether a more complex model improves on that, and whether it treats poorer neighborhoods the same way. The audience is transportation planners and analysts who prioritize safety spending, plus data scientists who might be tempted to build more complicated tools.</font>

### <font color='blue'>Methods</font>

<font color='blue'>Each row is a census tract in a target year from 2016 to 2024, and the outcome is the number of crashes that injured or killed a pedestrian. Features use only earlier years. History features include the previous three years of harm counts, their average and recent deaths. Neighborhood features are population, poverty rate, the shares of workers who walk or take transit, commercial share of floor area, density and borough. I compared two simple baselines, last year's count and the average of the last three years, with two stronger baselines, a five year average and a recency weighted average. I also fit gradient boosted models with a Poisson objective on history only, neighborhood only and both, and a linear Poisson regression. Models trained on target years 2016 to 2022 and were tested on 2023 and 2024, on tracts in ZIP clusters held out of training, so the test is out of time and out of space. I measured how much of the next year's harm falls in the 10 percent of tracts each method ranks highest. Intervals come from 1,000 bootstrap resamples of tracts. To audit fairness I measured, by poverty quartile, the share of the truly worst 10 percent of tracts that each method flagged (recall) and the share of its flags that were truly in that group (precision).</font>

### <font color='blue'>Results</font>

<font color='blue'>In the top 10 percent of ranked tracts, the three year average captured 25.3 percent of next-year harm (95% CI 23.9 to 26.5), the combined model 25.2 percent (23.9 to 26.7), and last year's count 24.0 percent (22.7 to 25.4). The linear model reached 23.7 percent (22.4 to 25.1), history only 22.5 percent (21.2 to 23.8) and neighborhood only 19.7 percent (18.2 to 21.1). Random ranking would capture 10 percent, and a perfect ranking 30.4 percent, because individual crashes are partly random. The difference between the combined model and the three year average is 0.0 points (interval -0.5 to +0.6), so the data cannot separate them and rule out a gain larger than about half a point. The five year average and the recency weighted average each captured 25.4 percent, within a tenth of a point of the three year average.</font>

<font color='blue'>Neighborhood characteristics are relevant but add nothing once history is known. A model using only neighborhood features roughly doubles random ranking, but when combined with history it improves the fit to counts (Poisson deviance 1.46 against 1.99 for history alone) mainly because it knows each tract's population. Shuffling features shows that the three year average and tract population matter most, followed by commercial share and last year's count. Poverty, walking, transit, density and borough contribute almost nothing. Shuffling splits credit among features that overlap, so the small values for single lag years and for poverty do not prove those variables carry no information. The ranking does match the earlier finding that poverty adds little once history and land use are known.</font>

<font color='blue'>The methods find high-harm tracts about equally well across poverty levels, and I found no sign that poorer tracts are found less well. For the three year average, recall of the worst 10 percent of tracts, from lowest to highest poverty quartile, was 0.57 (95% CI 0.40 to 0.68), 0.57 (0.43 to 0.65), 0.52 (0.43 to 0.64) and 0.62 (0.56 to 0.72). For the combined model it was 0.56 (0.39 to 0.68), 0.53 (0.40 to 0.63), 0.56 (0.45 to 0.66) and 0.62 (0.57 to 0.73). The gap between the highest and lowest quartile is +0.05 (-0.07 to +0.26) for the three year average and +0.07 (-0.05 to +0.28) for the combined model. These intervals rule out recall being lower in the poorest quartile by more than about 6 points, but they are wide, so they cannot exclude larger differences in the other direction. Precision, the share of flagged tracts that truly belong to the worst group, was 0.58, 0.62, 0.61 and 0.54 across quartiles for the three year average (0.54 has an interval of 0.48 to 0.62, and the lowest quartile 0.58 has 0.42 to 0.71). About 46 percent of flags in the highest poverty quartile missed the worst group, against 42 percent in the lowest, though the intervals overlap heavily. Predictions run below actual counts in every quartile for the history based methods, at 83 to 94 percent of actual, mostly because the averages include the quieter pandemic years.</font>

### <font color='blue'>What this means for planners</font>

<font color='blue'>Where crash records exist, a trailing average of recent years is a strong guide to next year's high-harm tracts. Neighborhood models could still help where history is thin or unreliable. The ceiling is low even for a perfect ranking, so any list of priority tracts will miss most of next year's harm, which fits my earlier finding that harm is spread across the city.</font>

### <font color='blue'>Limits</font>

<font color='blue'>The neighborhood features come from a single 2019 to 2023 snapshot applied to every year, so earlier years pair with later neighborhood data. The training period includes the pandemic. I did not tune the models, so a tuned version might narrow or widen the gaps. Whole ZIP clusters were held out, which limits leakage between neighbors only if the clusters separate well. The outcome counts crashes with a harmed pedestrian, not their severity, and deaths are too rare for a reliable death model. The fairness audit has limited power, since each poverty quartile holds only a few dozen of the worst tracts per year, so its intervals on recall span 0.2 to 0.3.</font>

### <font color='blue'>Ideas for future work</font>

<font color='blue'>I think future work could see an empirical Bayes model that blends a regression expectation with each tract's own history, which could reduce noise. Adding neighboring tracts' counts would also capture corridors that cross boundaries. Predicting at the street segment or intersection level would take road design into consideration, and visual models on imagery could analyze what an intersection looks like. I did a coverage check of Mapillary street images and found images at 87 percent of the 100 worst pedestrian sites but only 63 percent of 100 ordinary harmed sites, so image availability would need controlling in any vision model.</font>

## <font color="blue">Additional Work: Exposure-Adjusted Pedestrian Harm</font>

<font color="blue">This section extends the Milestone 5 analysis after the milestones were complete. It asks whether the link between neighborhood poverty and pedestrian harm holds once we account for how many people walk and for the design of the street.</font>

### <font color="blue">Overview</font>

<font color="blue">**Question.** Does neighborhood poverty relate to pedestrian harm after we account for pedestrian exposure and street design, and does the answer depend on how the city is divided into zones?</font>

<font color="blue">**Approach.**</font>
1. <font color="blue">Take modeled peak-hour pedestrian volumes from the NYC Walks model, which covers every sidewalk, crosswalk and footpath in the city.</font>
2. <font color="blue">Infer crosswalks as the pedestrian segments that cross a street centerline, then group them into sites, which are mostly intersections.</font>
3. <font color="blue">Assign each pedestrian harm crash from 2019 through 2023 to the nearest site within 30 meters.</font>
4. <font color="blue">Model crashes per site against crossing volume, street width, travel lanes, speed limit and tract poverty, with standard errors clustered by ZIP area.</font>
5. <font color="blue">Test how fragile the result is. The checks cover the estimator, the site definition, time, borough, measurement error in the volumes, pedestrian deaths and geographic scale.</font>

### <font color="blue">Headline results</font>

<font color="blue">Poverty rate ratio per 10 point increase in the tract poverty rate, with 95 percent intervals.</font>

| <font color="blue">Analysis</font> | <font color="blue">Rate ratio</font> |
|---|---|
| <font color="blue">Tracts, residents as exposure (Milestone 5)</font> | <font color="blue">1.05 (1.02 to 1.08)</font> |
| <font color="blue">Tracts, population and crossing volume as covariates</font> | <font color="blue">1.11 (1.08 to 1.15)</font> |
| <font color="blue">Sites, crossing volume only (model A)</font> | <font color="blue">1.30 (1.26 to 1.35)</font> |
| <font color="blue">Sites, plus street width, lanes and speed (model B)</font> | <font color="blue">1.28 (1.23 to 1.32)</font> |
| <font color="blue">Sites, plus tract controls (model C)</font> | <font color="blue">1.15 (1.11 to 1.18)</font> |

| <font color="blue">Pedestrian deaths at sites (model D)</font> | <font color="blue">1.29 (1.17 to 1.42)</font> |

- <font color="blue">**Robustness.** The site level ratio stays between 1.12 and 1.25 across alternative estimators, site definitions, time periods, boroughs, measurement error in the volumes and geographic scales.</font>
- <font color="blue">**Scale.** With exposure and street design, ratios run from 1.13 for tracts to 1.20 for ZIP areas. With residents only, they stay between 1.06 and 1.09. Exposure drives the estimate more than geography does.</font>
- <font color="blue">**Deaths.** The ratio is 1.29, and checks range from 1.23 to 1.35. About 60 percent of the gap comes from more harm crashes (1.16) and about 40 percent from a higher chance that a crash is fatal (1.11, interval 1.00 to 1.23).</font>

### <font color="blue">How to read the results</font>

<font color="blue">The unadjusted gap (1.28 to 1.30) shows how much more pedestrian harm poorer areas carry. The adjusted gap (1.16) shows how much remains after measured exposure, street design and tract context. Transit and walking shares may sit on the causal path, so the two ratios answer different questions. All results show association, not cause.</font>

#### <font color="blue">Step 3. Infer crosswalks and build tract crossing exposure</font>

````python
# Step 2 of the crossing exposure build.
# The NYC Walks file has no segment type, so we infer crosswalks as the segments that cross a
# street centerline. Sidewalks run alongside streets and crosswalks cross them. Park footpaths
# do not cross roadbeds, so they drop out of this measure.
# Run after step 1, which saved the centerline file to Drive.

# %% C2. Infer crosswalks and build tract level crossing exposure
import os
import zipfile
import numpy as np
import pandas as pd
import geopandas as gpd
import requests
from google.colab import drive

drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
MEMBER = 'NYC_pednetwork_estimates_counts_2018-2019.geojson'
LOCAL = f'/content/{MEMBER}'

if not os.path.exists(LOCAL):
    with zipfile.ZipFile(f'{DRIVE_DIR}/supp_data_1_network.zip') as z:
        with z.open(MEMBER) as src, open(LOCAL, 'wb') as dst:
            while True:
                chunk = src.read(1 << 20)
                if not chunk:
                    break
                dst.write(chunk)

TRACT_ZIP = '/content/ny_tracts2020.zip'
if not os.path.exists(TRACT_ZIP):
    url = 'https://www2.census.gov/geo/tiger/GENZ2020/shp/cb_2020_36_tract_500k.zip'
    with open(TRACT_ZIP, 'wb') as f:
        f.write(requests.get(url, timeout=120).content)

NYC_COUNTIES = {'005': 'Bronx', '047': 'Brooklyn', '061': 'Manhattan',
                '081': 'Queens', '085': 'Staten Island'}
tracts = gpd.read_file(TRACT_ZIP)
tracts = tracts[tracts['COUNTYFP'].isin(NYC_COUNTIES)][['GEOID', 'geometry']].to_crs(6538)

# Street centerlines
cl = gpd.read_file(f'{DRIVE_DIR}/centerline_inkn-q76z.geojson').to_crs(6538)
for c in ['rw_type', 'status', 'streetwidth', 'posted_speed']:
    cl[c] = pd.to_numeric(cl[c], errors='coerce')

# Read the example names to interpret each road type code before trusting it
summary = cl.groupby('rw_type').agg(
    n=('physicalid', 'size'),
    median_width_ft=('streetwidth', 'median'),
    share_nonvehicular=('trafdir', lambda s: (s == 'NV').mean()))
summary['examples'] = cl.groupby('rw_type')['full_street_name'].agg(
    lambda s: ' | '.join(s.dropna().drop_duplicates().head(4)))
print('Road types in the centerline file')
print(summary.round(2).to_string())
print('\nnonped values')
print(cl['nonped'].value_counts(dropna=False).head(8).to_string())

# Pedestrian volumes
WKDY = ['predwkdyAM', 'predwkdyMD', 'predwkdyPM']
WKND = ['predwkndAM', 'predwkndMD', 'predwkndPM']
seg = gpd.read_file(LOCAL, engine='pyogrio', columns=WKDY + WKND).to_crs(6538)
seg[WKDY + WKND] = seg[WKDY + WKND].fillna(0).clip(lower=0)
seg['len_m'] = seg.geometry.length
seg['vol_wkdy'] = seg[WKDY].mean(axis=1)
seg['vol_wknd'] = seg[WKND].mean(axis=1)
seg['vol_peak'] = seg[WKDY + WKND].mean(axis=1)
seg['seg_id'] = np.arange(len(seg))

# Start with ordinary streets only. Add other codes here only after reading the table above.
STREET_TYPES = [1]
road = cl[(cl['status'] == 2) & (cl['rw_type'].isin(STREET_TYPES)) & (cl['trafdir'] != 'NV')]
road = road[['physicalid', 'streetwidth', 'posted_speed', 'geometry']]
print('\nstreet centerlines used:', len(road))

hit = gpd.sjoin(seg[['seg_id', 'geometry']], road, how='inner', predicate='intersects')
n_hit = hit.groupby('seg_id').size()
widest = (hit.sort_values('streetwidth', ascending=False, na_position='last')
          .drop_duplicates('seg_id')[['seg_id', 'streetwidth', 'posted_speed']])
seg['n_streets'] = seg['seg_id'].map(n_hit).fillna(0).astype(int)
seg = seg.merge(widest, on='seg_id', how='left')
seg['is_cross'] = seg['n_streets'] > 0

print('segments flagged as crossings:', int(seg['is_cross'].sum()),
      f"({seg['is_cross'].mean():.1%} of all)")
print('flagged segments that meet two or more streets:', int((seg['n_streets'] >= 2).sum()))
print('\nMedian length and peak volume, crossings versus other segments')
print(seg.groupby('is_cross')[['len_m', 'vol_peak']].median().round(1).to_string())

# Assign each crossing to the tract that holds its midpoint
cr = seg[seg['is_cross']].copy()
cr['geometry'] = cr.geometry.interpolate(0.5, normalized=True)
cr = gpd.GeoDataFrame(cr, geometry='geometry', crs=seg.crs)
keep = ['vol_wkdy', 'vol_wknd', 'vol_peak', 'streetwidth', 'posted_speed', 'geometry']
cj = gpd.sjoin(cr[keep], tracts, how='left', predicate='within')
print('crossings without a tract:', int(cj['GEOID'].isna().sum()))
cj = cj.dropna(subset=['GEOID'])

cj['vol_x_spd'] = cj['vol_peak'] * cj['posted_speed']
cj['vol_spd'] = cj['vol_peak'].where(cj['posted_speed'].notna(), 0)
cj['vol_x_wid'] = cj['vol_peak'] * cj['streetwidth']
cj['vol_wid'] = cj['vol_peak'].where(cj['streetwidth'].notna(), 0)

agg = (cj.groupby('GEOID')
       .agg(n_crossings=('vol_peak', 'size'),
            cross_vol_wkdy=('vol_wkdy', 'sum'),
            cross_vol_wknd=('vol_wknd', 'sum'),
            cross_vol_peak=('vol_peak', 'sum'),
            _xs=('vol_x_spd', 'sum'), _vs=('vol_spd', 'sum'),
            _xw=('vol_x_wid', 'sum'), _vw=('vol_wid', 'sum'))
       .reset_index())
agg['wtd_speed_mph'] = agg['_xs'] / agg['_vs'].replace(0, np.nan)
agg['wtd_street_width_ft'] = agg['_xw'] / agg['_vw'].replace(0, np.nan)
agg = agg.drop(columns=['_xs', '_vs', '_xw', '_vw'])
agg['mean_cross_vol'] = agg['cross_vol_peak'] / agg['n_crossings']
agg['county'] = agg['GEOID'].str[2:5].map(NYC_COUNTIES)

out_csv = f'{DRIVE_DIR}/tract_crossing_exposure_2018_2019_volumes.csv'
agg.to_csv(out_csv, index=False)
print('\ntracts with crossings:', len(agg), 'of', len(tracts), '| saved to', out_csv)

cols = ['GEOID', 'county', 'n_crossings', 'cross_vol_peak', 'mean_cross_vol',
        'wtd_speed_mph', 'wtd_street_width_ft']
print('\nTop 10 tracts by crossing volume')
print(agg.sort_values('cross_vol_peak', ascending=False).head(10)[cols].round(1).to_string(index=False))
print('\nMedian crossing volume by borough')
print(agg.groupby('county')['cross_vol_peak'].median().round(0).to_string())
print('\nDistribution')
print(agg[['n_crossings', 'cross_vol_peak', 'wtd_speed_mph', 'wtd_street_width_ft']]
      .describe().round(1).to_string())

# How different is this from the first exposure measure?
old_csv = f'{DRIVE_DIR}/tract_exposure_2018_2019_volumes.csv'
if os.path.exists(old_csv):
    old = pd.read_csv(old_csv, dtype={'GEOID': str})
    both = old.merge(agg, on='GEOID')
    rho = both['ped_m_peak'].corr(both['cross_vol_peak'], method='spearman')
    print(f'\nSpearman correlation with the first measure (pedestrian meters): {rho:.2f}')
````

#### <font color="blue">Step 4. Merge exposure onto the tract table and refit the poverty models</font>

````python
# Merge crossing exposure onto the Milestone 5 tract table and refit the poverty models.
# Run in the TDSP Research Sandbox after the Milestone 1 and Milestone 5 cells.
# Uses these names from your session: tract_table, HEADLINE_YEARS, CONTROLS, tracts, smf, np, pd.

# %% E1. Merge exposure, check that the volumes are plausible, then fit the models
import warnings
import numpy as np
import pandas as pd

warnings.filterwarnings('ignore')

DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
expo = pd.read_csv(f'{DRIVE_DIR}/tract_crossing_exposure_2018_2019_volumes.csv', dtype={'GEOID': str})

base = tract_table(HEADLINE_YEARS, 'local')
keep = ['GEOID', 'county', 'n_crossings', 'cross_vol_peak', 'mean_cross_vol',
        'wtd_speed_mph', 'wtd_street_width_ft']
m = base.merge(expo[keep], on='GEOID', how='inner')
print(f'tracts in the Milestone 5 table: {len(base)}, with crossing exposure: {len(m)}')

tr = tracts.to_crs(6538)
tr['area_km2'] = tr.geometry.area / 1e6
m = m.merge(tr[['GEOID', 'area_km2']], on='GEOID', how='left')
m['density'] = m['Total_Population'] / m['area_km2']
m['log_pop'] = np.log(m['Total_Population'])
m['log_cross'] = np.log(m['cross_vol_peak'])

# 1. Are the modeled volumes plausible? Compare them with density and ACS walking share.
print('\n1. Plausibility of the modeled volumes')
by = m.groupby('county').agg(
    tracts=('GEOID', 'size'),
    median_crossings=('n_crossings', 'median'),
    median_vol_per_crossing=('mean_cross_vol', 'median'),
    median_walk_share=('Walk_Share', 'median'),
    median_density=('density', 'median'))
print(by.round(1).to_string())
for col in ['density', 'Walk_Share', 'Poverty_Rate_Percent']:
    rho = m['mean_cross_vol'].corr(m[col], method='spearman')
    print(f'Spearman, volume per crossing vs {col}: {rho:.2f}')
rho = m['cross_vol_peak'].corr(m['Poverty_Rate_Percent'], method='spearman')
print(f'Spearman, total crossing volume vs poverty rate: {rho:.2f}')


# 2. Models. The exposure function lets each specification choose its own denominator.
def run(d, outcome, terms, expo_fn):
    rhs = ' + '.join(['Poverty_Rate_Percent'] + terms)
    model = smf.negativebinomial(f'{outcome} ~ {rhs}', data=d, exposure=expo_fn(d))
    groups = pd.factorize(d['ZCTA_cluster'])[0]
    return model.fit(disp=0, maxiter=500, cov_type='cluster', cov_kwds={'groups': groups})


def rr(fit, term='Poverty_Rate_Percent'):
    lo, hi = fit.conf_int().loc[term]
    return f'{np.exp(10 * fit.params[term]):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})'


def coef(fit, term):
    lo, hi = fit.conf_int().loc[term]
    return f'{fit.params[term]:.2f} ({lo:.2f} to {hi:.2f})'


pop_exp = lambda d: d['Total_Population'] * d['n_years']
cross_exp = lambda d: d['cross_vol_peak'] * d['n_years']
years_exp = lambda d: d['n_years']
no_si = m[m['county'] != 'Staten Island'].copy()
both = CONTROLS + ['log_pop', 'log_cross']

specs = [
    ('1 Residents as exposure, with controls (current)', m, CONTROLS, pop_exp),
    ('2 Residents as exposure, no controls', m, [], pop_exp),
    ('3 Crossing volume as exposure, with controls', m, CONTROLS, cross_exp),
    ('4 Crossing volume as exposure, no controls', m, [], cross_exp),
    ('5 Both as covariates, with controls', m, both, years_exp),
    ('6 Same as 5, plus borough effects', m, both + ['C(county)'], years_exp),
    ('7 Same as 5, Staten Island excluded', no_si, both, years_exp),
]

rows = []
for label, d, terms, expo_fn in specs:
    row = {'model': label, 'tracts': len(d)}
    for outcome, name in [('Ped_Injured', 'injuries'), ('Ped_Killed', 'deaths')]:
        try:
            fit = run(d, outcome, terms, expo_fn)
            row[f'{name}_poverty_RR_per_10'] = rr(fit)
            if 'log_cross' in terms:
                row[f'{name}_elasticity_pop'] = coef(fit, 'log_pop')
                row[f'{name}_elasticity_cross'] = coef(fit, 'log_cross')
        except Exception as e:
            row[f'{name}_poverty_RR_per_10'] = f'failed: {str(e)[:40]}'
    rows.append(row)

res = pd.DataFrame(rows)
pd.set_option('display.width', 250)
pd.set_option('display.max_colwidth', 60)
print('\n2. Poverty rate ratio per +10 points, 95% interval, clustered by ZIP area')
print(res[['model', 'tracts', 'injuries_poverty_RR_per_10', 'deaths_poverty_RR_per_10']].to_string(index=False))
print('\nElasticities in the models with both covariates (a value below 1 means risk per walker falls as volume rises)')
cols = [c for c in res.columns if 'elasticity' in c]
print(res.loc[res[cols[0]].notna(), ['model'] + cols].to_string(index=False))

res.to_csv(f'{DRIVE_DIR}/exposure_model_results.csv', index=False)
print('\nsaved to', f'{DRIVE_DIR}/exposure_model_results.csv')
````

#### <font color="blue">Step 5. Build intersection sites and fit the first site level models</font>

````python
# Crossing site level model of pedestrian harm crashes, 2019 to 2023.
# Run in the TDSP Research Sandbox after the Milestone 1 and Milestone 5 cells.
# Uses these names from your session: crashes, tracts, tract_table, HEADLINE_YEARS, smf, gpd.
#
# Design
#  1. Infer crosswalks as pedestrian segments that cross a street centerline (as before).
#  2. Group crosswalk midpoints within 25 m of each other into sites, which are mostly intersections.
#     Crash coordinates sit at the intersection center, so a site is the right unit, not one crosswalk.
#  3. Count pedestrian harm crashes within 30 m of each site's center.
#  4. Model counts with site crossing volume, street width, lanes, speed and tract poverty.

# %% F1. Build sites, assign crashes, fit the models
import os
import zipfile
import warnings
import numpy as np
import pandas as pd
import geopandas as gpd
import statsmodels.api as sm
from sklearn.cluster import DBSCAN
from google.colab import drive

warnings.filterwarnings('ignore')
drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
MEMBER = 'NYC_pednetwork_estimates_counts_2018-2019.geojson'
LOCAL = f'/content/{MEMBER}'
COUNTY_NAMES = {'005': 'Bronx', '047': 'Brooklyn', '061': 'Manhattan', '081': 'Queens', '085': 'Staten Island'}
SITE_EPS_M = 25      # crosswalks closer than this form one site
CRASH_RADIUS_M = 30  # a crash belongs to the nearest site within this distance

if not os.path.exists(LOCAL):
    with zipfile.ZipFile(f'{DRIVE_DIR}/supp_data_1_network.zip') as z:
        with z.open(MEMBER) as src, open(LOCAL, 'wb') as dst:
            while True:
                chunk = src.read(1 << 20)
                if not chunk:
                    break
                dst.write(chunk)

# Street centerlines, ordinary streets only
cl = gpd.read_file(f'{DRIVE_DIR}/centerline_inkn-q76z.geojson').to_crs(6538)
for c in ['rw_type', 'status', 'streetwidth', 'posted_speed', 'number_travel_lanes']:
    cl[c] = pd.to_numeric(cl[c], errors='coerce')
road = cl[(cl['status'] == 2) & (cl['rw_type'] == 1) & (cl['trafdir'] != 'NV')]
road = road[['streetwidth', 'posted_speed', 'number_travel_lanes', 'geometry']]

# Pedestrian volumes and crosswalk inference
WKDY = ['predwkdyAM', 'predwkdyMD', 'predwkdyPM']
WKND = ['predwkndAM', 'predwkndMD', 'predwkndPM']
seg = gpd.read_file(LOCAL, engine='pyogrio', columns=WKDY + WKND).to_crs(6538)
seg[WKDY + WKND] = seg[WKDY + WKND].fillna(0).clip(lower=0)
seg['vol_peak'] = seg[WKDY + WKND].mean(axis=1)
seg['seg_id'] = np.arange(len(seg))

hit = gpd.sjoin(seg[['seg_id', 'geometry']], road, how='inner', predicate='intersects')
widest = (hit.sort_values('streetwidth', ascending=False, na_position='last')
          .drop_duplicates('seg_id')
          [['seg_id', 'streetwidth', 'posted_speed', 'number_travel_lanes']])
cross = seg.merge(widest, on='seg_id', how='inner')
print('crosswalk segments:', len(cross))

# Group crosswalks into sites
mid = cross.geometry.interpolate(0.5, normalized=True)
xy = np.c_[mid.x.values, mid.y.values]
cross['site'] = DBSCAN(eps=SITE_EPS_M, min_samples=1).fit_predict(xy)
cross['x'], cross['y'] = xy[:, 0], xy[:, 1]

sites = (cross.groupby('site')
         .agg(n_cross=('seg_id', 'size'), site_vol=('vol_peak', 'sum'),
              max_width=('streetwidth', 'max'), max_speed=('posted_speed', 'max'),
              max_lanes=('number_travel_lanes', 'max'), x=('x', 'mean'), y=('y', 'mean'))
         .reset_index())
site_pts = gpd.GeoDataFrame(sites[['site']], geometry=gpd.points_from_xy(sites['x'], sites['y']), crs=6538)
print('sites:', len(sites), '| median crosswalks per site:', int(sites['n_cross'].median()))

# Tract attributes from the Milestone 5 table
tm0 = tract_table(HEADLINE_YEARS, 'local')
tr = tracts.to_crs(6538)[['GEOID', 'geometry']]
js = gpd.sjoin(site_pts, tr, how='left', predicate='within')
js = js[~js.index.duplicated(keep='first')]
sites['GEOID'] = js['GEOID'].values
cols = ['GEOID', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share', 'Commercial_Share',
        'ZCTA_cluster', 'Total_Population']
sites = sites.merge(tm0[cols], on='GEOID', how='inner')
sites['county'] = sites['GEOID'].str[2:5].map(COUNTY_NAMES)
print('sites in Milestone 5 tracts:', len(sites))

# Pedestrian harm crashes, 2019 to 2023, highway crashes dropped
c = crashes[crashes['YEAR'].isin(HEADLINE_YEARS) & ~crashes['is_highway']].copy()
c['ped_inj'] = c['NUMBER OF PEDESTRIANS INJURED'].fillna(0)
c['ped_kill'] = c['NUMBER OF PEDESTRIANS KILLED'].fillna(0)
c = c[(c['ped_inj'] + c['ped_kill']) > 0]
c = c[c['LATITUDE'].between(40.4, 41.0) & c['LONGITUDE'].between(-74.3, -73.6)]
cp = gpd.GeoDataFrame(c[['ped_inj', 'ped_kill']].reset_index(drop=True),
                      geometry=gpd.points_from_xy(c['LONGITUDE'], c['LATITUDE']), crs=4326).to_crs(6538)
print('pedestrian harm crashes with coordinates:', len(cp))


def assign(radius):
    j = gpd.sjoin_nearest(cp, site_pts, how='left', max_distance=radius)
    j = j[~j.index.duplicated(keep='first')]
    ok = j['site'].notna()
    agg = (j[ok].groupby('site')
           .agg(harm_crashes=('site', 'size'), ped_injured=('ped_inj', 'sum'), ped_killed=('ped_kill', 'sum')))
    return agg, ok.mean()


def build(radius):
    agg, share = assign(radius)
    d = sites.merge(agg, left_on='site', right_index=True, how='left')
    d[['harm_crashes', 'ped_injured', 'ped_killed']] = d[['harm_crashes', 'ped_injured', 'ped_killed']].fillna(0)
    d['log_vol'] = np.log(d['site_vol'].clip(lower=1))
    d['width10'] = d['max_width'] / 10
    d['speed5'] = d['max_speed'] / 5
    d['lanes'] = d['max_lanes']
    d = d.dropna(subset=['log_vol', 'width10', 'speed5', 'lanes', 'Poverty_Rate_Percent', 'Walk_Share',
                         'Transit_Share', 'Commercial_Share', 'ZCTA_cluster'])
    return d, share


def fit(d, terms):
    rhs = ' + '.join(['Poverty_Rate_Percent'] + terms)
    groups = pd.factorize(d['ZCTA_cluster'])[0]
    try:
        model = smf.negativebinomial(f'harm_crashes ~ {rhs}', data=d)
        return model.fit(disp=0, maxiter=300, cov_type='cluster', cov_kwds={'groups': groups}), 'NB'
    except Exception:
        model = smf.glm(f'harm_crashes ~ {rhs}', data=d, family=sm.families.Poisson())
        return model.fit(cov_type='cluster', cov_kwds={'groups': groups}), 'Poisson'


def rr10(res):
    lo, hi = res.conf_int().loc['Poverty_Rate_Percent']
    return f"{np.exp(10 * res.params['Poverty_Rate_Percent']):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})"


d, share = build(CRASH_RADIUS_M)
print(f'\nShare of harm crashes within {CRASH_RADIUS_M} m of a site: {share:.1%}')
print(f'sites in the model: {len(d)} | sites with at least one harm crash: {(d["harm_crashes"] > 0).mean():.1%}'
      f' | mean crashes per site: {d["harm_crashes"].mean():.2f}')

design = ['log_vol', 'width10', 'lanes', 'speed5']
more = design + ['n_cross', 'Walk_Share', 'Transit_Share', 'Commercial_Share']
specs = [('A  crossing volume only', ['log_vol']),
         ('B  + street width, lanes, speed', design),
         ('C  + crosswalks per site and tract controls', more),
         ('D  + borough effects', more + ['C(county)'])]

rows, fits = [], {}
for label, terms in specs:
    res, kind = fit(d, terms)
    fits[label] = res
    rows.append({'model': label, 'estimator': kind, 'poverty_RR_per_10': rr10(res),
                 'volume_elasticity': (f"{res.params['log_vol']:.2f}" if 'log_vol' in res.params else '')})
print('\nPoverty rate ratio per +10 points, 95% interval, clustered by ZIP area')
print(pd.DataFrame(rows).to_string(index=False))

full = fits[specs[-1][0]]
ci = full.conf_int()
tab = pd.DataFrame({'coef': full.params, 'lo': ci[0], 'hi': ci[1], 'p': full.pvalues}).round(3)
print('\nFull model coefficients. Units are log counts per unit of each variable.')
print('width10 is per 10 feet, speed5 per 5 mph, lanes per travel lane, Poverty_Rate_Percent per point.')
print(tab.to_string())

print('\nSensitivity to the crash assignment radius, full model')
for radius in [20, 30, 50]:
    dr, sh = build(radius)
    res, kind = fit(dr, more + ['C(county)'])
    print(f'  {radius} m: {sh:.1%} of crashes assigned, poverty RR per +10 = {rr10(res)} ({kind})')

print('\nTen sites with the most harm crashes')
show = ['site', 'county', 'n_cross', 'site_vol', 'max_width', 'max_lanes', 'max_speed', 'harm_crashes', 'Poverty_Rate_Percent']
print(d.sort_values('harm_crashes', ascending=False).head(10)[show].round(1).to_string(index=False))

d.to_csv(f'{DRIVE_DIR}/crossing_sites_model_data.csv', index=False)
print('\nsaved to', f'{DRIVE_DIR}/crossing_sites_model_data.csv')
````

#### <font color="blue">Step 6. Refit with robust estimators and run the first checks</font>

````python
# Robust refit of the crossing site models, reading the table saved by the previous cell.
# Poisson with cluster robust standard errors is the primary estimator. It stays consistent under
# overdispersion, and the negative binomial GLM with a moment based alpha is a cross check.
# Run in the Colab sandbox or any session with Drive mounted. Needs statsmodels.

# %% G1. Site level models, robust estimators
import warnings
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from google.colab import drive

warnings.filterwarnings('ignore')
drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
df = pd.read_csv(f'{DRIVE_DIR}/crossing_sites_model_data.csv', dtype={'GEOID': str})
print(df.shape)

for c in ['Walk_Share', 'Transit_Share', 'Commercial_Share', 'n_cross']:
    df[c + '_z'] = (df[c] - df[c].mean()) / df[c].std()
groups = pd.factorize(df['ZCTA_cluster'])[0]

design = 'log_vol + width10 + lanes + speed5'
ctrl = ' + n_cross_z + Walk_Share_z + Transit_Share_z + Commercial_Share_z'
specs = {
    'A volume only': 'log_vol',
    'B + street design': design,
    'C + crosswalks per site, tract controls': design + ctrl,
    'D + borough effects': design + ctrl + ' + C(county)',
}


def rr(res, term='Poverty_Rate_Percent'):
    lo, hi = res.conf_int().loc[term]
    return f"{np.exp(10 * res.params[term]):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})"


def fit_pois(f, data, groups_, y='harm_crashes'):
    model = smf.glm(f'{y} ~ Poverty_Rate_Percent + {f}', data=data, family=sm.families.Poisson())
    return model.fit(cov_type='cluster', cov_kwds={'groups': groups_})


def fit_nbglm(pres, f, data, groups_, y='harm_crashes'):
    mu = np.asarray(pres.mu)
    yy = data[y].values
    ystar = ((yy - mu) ** 2 - yy) / mu
    alpha = max(float((mu * ystar).sum() / (mu * mu).sum()), 1e-6)
    model = smf.glm(f'{y} ~ Poverty_Rate_Percent + {f}', data=data,
                    family=sm.families.NegativeBinomial(alpha=alpha))
    return model.fit(cov_type='cluster', cov_kwds={'groups': groups_}), alpha


rows, pfits = [], {}
for k, f in specs.items():
    p = fit_pois(f, df, groups)
    pfits[k] = p
    nbres, alpha = fit_nbglm(p, f, df, groups)
    rows.append({'model': k, 'Poisson_robust_RR': rr(p), 'NB_RR': rr(nbres),
                 'alpha_hat': round(alpha, 2), 'vol_elasticity': round(p.params['log_vol'], 2)})
print('\nPoverty rate ratio per +10 points, cluster robust by ZIP area')
print(pd.DataFrame(rows).to_string(index=False))

pD = pfits['D + borough effects']
ci = pD.conf_int()
tab = pd.DataFrame({'RR': np.exp(pD.params), 'lo': np.exp(ci[0]), 'hi': np.exp(ci[1]), 'p': pD.pvalues}).round(3)
print('\nModel D rate ratios per unit. _z terms are per 1 SD, width10 per 10 ft, speed5 per 5 mph.')
print(tab.drop(index='Intercept').to_string())
print('log_vol elasticity:', round(pD.params['log_vol'], 2))

lnA = pfits['A volume only'].params['Poverty_Rate_Percent']
print('\nHow much of the poverty gap each block of controls absorbs')
for k in specs:
    b = pfits[k].params['Poverty_Rate_Percent']
    print(f"  {k}: log RR per +10 = {10 * b:.3f} ({100 * (1 - b / lnA):.0f}% smaller than A)")

# Robustness
cut = df['harm_crashes'].quantile(0.99)
sub = df[df['harm_crashes'] <= cut].copy()
g2 = pd.factorize(sub['ZCTA_cluster'])[0]
fD = specs['D + borough effects']
print(f'\nRobustness for model D (top 1% cutoff is {cut:.0f} crashes)')
print('  drop top 1% of sites:', rr(fit_pois(fD, sub, g2)), '| sites:', len(sub))
print('  pedestrians injured:', rr(fit_pois(fD, df, groups, y='ped_injured')))
print('  pedestrians killed:', rr(fit_pois(fD, df, groups, y='ped_killed')),
      '| deaths in sample:', int(df['ped_killed'].sum()))

# Does the poverty gap change with volume or street width?
dd = df.copy()
dd['vol_c'] = dd['log_vol'] - dd['log_vol'].mean()
dd['wid_c'] = dd['width10'] - dd['width10'].mean()
fi = fD + ' + Poverty_Rate_Percent:vol_c + Poverty_Rate_Percent:wid_c'
pi = smf.glm(f'harm_crashes ~ Poverty_Rate_Percent + {fi}', data=dd,
             family=sm.families.Poisson()).fit(cov_type='cluster', cov_kwds={'groups': groups})
for t in ['Poverty_Rate_Percent:vol_c', 'Poverty_Rate_Percent:wid_c']:
    lo, hi = pi.conf_int().loc[t]
    print(f'  interaction {t}: coef {pi.params[t]:.4f} ({lo:.4f} to {hi:.4f}), p = {pi.pvalues[t]:.3f}')
````

#### <font color="blue">Step 7. How much could error in the pedestrian volumes matter?</font>

````python
# Measurement error sensitivity for the site level model (SIMEX).
# Question: how much could error in the modeled pedestrian volumes move the poverty estimate?
# Needs only crossing_sites_model_data.csv on Drive. Run it in any session with Drive mounted.
#
# Method. Treat observed log volume as true log volume plus classical noise. For an assumed
# reliability (share of variance that is real signal), add extra noise at several multiples of the
# estimated error variance, refit, and extrapolate the coefficients back to zero error.
# Classical error is a worst case. Model predictions often behave more like Berkson error, which
# does not bias the volume coefficient, so read the results as a plausible range, not a correction.

# %% H1. SIMEX sensitivity
import time
import numpy as np
import pandas as pd
import patsy
from google.colab import drive

drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
df = pd.read_csv(f'{DRIVE_DIR}/crossing_sites_model_data.csv', dtype={'GEOID': str})
for c in ['Walk_Share', 'Transit_Share', 'Commercial_Share', 'n_cross']:
    df[c + '_z'] = (df[c] - df[c].mean()) / df[c].std()

formula = ('harm_crashes ~ Poverty_Rate_Percent + log_vol + width10 + lanes + speed5'
           ' + n_cross_z + Walk_Share_z + Transit_Share_z + Commercial_Share_z + C(county)')
yS, XS = patsy.dmatrices(formula, df, return_type='dataframe')
yv = yS.iloc[:, 0].values.astype(float)
Xv = XS.values.astype(float)
names = list(XS.columns)
i_pov, i_vol = names.index('Poverty_Rate_Percent'), names.index('log_vol')
var_obs = df['log_vol'].var()
t0 = time.time()


def pois_irls(Xm, beta0, iters=30, tol=1e-9):
    beta = beta0.copy()
    for _ in range(iters):
        eta = Xm @ beta
        mu = np.exp(np.clip(eta, -20, 20))
        z = eta + (yv - mu) / mu
        XtW = Xm.T * mu
        new = np.linalg.solve(XtW @ Xm, XtW @ z)
        done = np.max(np.abs(new - beta)) < tol
        beta = new
        if done:
            break
    return beta


b0 = pois_irls(Xv, np.zeros(Xv.shape[1]), iters=80)
b0 = pois_irls(Xv, b0)
naive = (b0[i_pov], b0[i_vol])
print(f'corr(log volume, poverty) across sites: {df["log_vol"].corr(df["Poverty_Rate_Percent"]):.3f}')
print(f'naive fit: poverty RR per +10 = {np.exp(10 * naive[0]):.3f}, volume elasticity = {naive[1]:.2f}')

rng = np.random.default_rng(2026)
LAMS, B = [0.5, 1.0, 1.5, 2.0], 12
rows = []
for rel in [0.95, 0.9, 0.8, 0.7]:
    s2 = (1 - rel) * var_obs
    pts = {0.0: naive}
    for lam in LAMS:
        acc = []
        for _ in range(B):
            Xs = Xv.copy()
            Xs[:, i_vol] = Xv[:, i_vol] + rng.normal(0, np.sqrt(lam * s2), len(Xv))
            bs = pois_irls(Xs, b0)
            acc.append((bs[i_pov], bs[i_vol]))
        pts[lam] = tuple(np.mean(acc, axis=0))
    L = np.array(sorted(pts))
    pov = np.array([pts[k][0] for k in L])
    vol = np.array([pts[k][1] for k in L])
    rows.append({
        'assumed reliability': rel,
        'poverty RR naive': round(np.exp(10 * naive[0]), 3),
        'poverty RR SIMEX quadratic': round(np.exp(10 * np.polyval(np.polyfit(L, pov, 2), -1.0)), 3),
        'poverty RR SIMEX linear': round(np.exp(10 * np.polyval(np.polyfit(L, pov, 1), -1.0)), 3),
        'elasticity naive': round(naive[1], 2),
        'elasticity SIMEX quadratic': round(np.polyval(np.polyfit(L, vol, 2), -1.0), 2),
    })
    print(f'reliability {rel} done after {time.time() - t0:.0f} s')

print('\nPoverty rate ratio per +10 points and volume elasticity after correcting for assumed error')
print(pd.DataFrame(rows).to_string(index=False))
````

````python
# Robustness checks for the site level poverty estimate.
#  1. Site definition: group crosswalks within 15, 25 or 40 m.
#  2. Time: 2019 to 2021 versus 2022 to 2023, and each year, with the same 2018 and 2019 volumes.
#  3. Leave one borough out.
# Run in the TDSP Research Sandbox after the Milestone 1 and Milestone 5 cells.
# Uses these names from the session: crashes, tracts, tract_table, HEADLINE_YEARS.
# Allow several minutes. The first part rebuilds the crosswalks and reads a 275 MB file.

# %% I1. Robustness checks
import os
import zipfile
import warnings
import numpy as np
import pandas as pd
import geopandas as gpd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from sklearn.cluster import DBSCAN
from google.colab import drive

warnings.filterwarnings('ignore')
drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
MEMBER = 'NYC_pednetwork_estimates_counts_2018-2019.geojson'
LOCAL = f'/content/{MEMBER}'
COUNTY_NAMES = {'005': 'Bronx', '047': 'Brooklyn', '061': 'Manhattan', '081': 'Queens', '085': 'Staten Island'}
RADIUS_M = 30
YEARS = sorted(HEADLINE_YEARS)

if not os.path.exists(LOCAL):
    with zipfile.ZipFile(f'{DRIVE_DIR}/supp_data_1_network.zip') as z:
        with z.open(MEMBER) as src, open(LOCAL, 'wb') as dst:
            while True:
                chunk = src.read(1 << 20)
                if not chunk:
                    break
                dst.write(chunk)

# Crosswalks, built once
cl = gpd.read_file(f'{DRIVE_DIR}/centerline_inkn-q76z.geojson').to_crs(6538)
for c in ['rw_type', 'status', 'streetwidth', 'posted_speed', 'number_travel_lanes']:
    cl[c] = pd.to_numeric(cl[c], errors='coerce')
road = cl[(cl['status'] == 2) & (cl['rw_type'] == 1) & (cl['trafdir'] != 'NV')]
road = road[['streetwidth', 'posted_speed', 'number_travel_lanes', 'geometry']]

WKDY = ['predwkdyAM', 'predwkdyMD', 'predwkdyPM']
WKND = ['predwkndAM', 'predwkndMD', 'predwkndPM']
seg = gpd.read_file(LOCAL, engine='pyogrio', columns=WKDY + WKND).to_crs(6538)
seg[WKDY + WKND] = seg[WKDY + WKND].fillna(0).clip(lower=0)
seg['vol_peak'] = seg[WKDY + WKND].mean(axis=1)
seg['seg_id'] = np.arange(len(seg))
hit = gpd.sjoin(seg[['seg_id', 'geometry']], road, how='inner', predicate='intersects')
widest = (hit.sort_values('streetwidth', ascending=False, na_position='last')
          .drop_duplicates('seg_id')[['seg_id', 'streetwidth', 'posted_speed', 'number_travel_lanes']])
cross = seg.merge(widest, on='seg_id', how='inner')
mid = cross.geometry.interpolate(0.5, normalized=True)
xy = np.c_[mid.x.values, mid.y.values]
print('crosswalk segments:', len(cross))

# Tract attributes and crash points
tm0 = tract_table(HEADLINE_YEARS, 'local')
cols = ['GEOID', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share', 'Commercial_Share', 'ZCTA_cluster']
tr = tracts.to_crs(6538)[['GEOID', 'geometry']]

c = crashes[crashes['YEAR'].isin(HEADLINE_YEARS) & ~crashes['is_highway']].copy()
c['ped_inj'] = c['NUMBER OF PEDESTRIANS INJURED'].fillna(0)
c['ped_kill'] = c['NUMBER OF PEDESTRIANS KILLED'].fillna(0)
c = c[(c['ped_inj'] + c['ped_kill']) > 0]
c = c[c['LATITUDE'].between(40.4, 41.0) & c['LONGITUDE'].between(-74.3, -73.6)]
cp = gpd.GeoDataFrame(c[['ped_inj', 'ped_kill']].assign(year=c['YEAR'].values).reset_index(drop=True),
                      geometry=gpd.points_from_xy(c['LONGITUDE'], c['LATITUDE']), crs=4326).to_crs(6538)
print('pedestrian harm crashes:', len(cp))


def make_dataset(eps):
    lab = DBSCAN(eps=eps, min_samples=1).fit_predict(xy)
    cr = pd.DataFrame({'site': lab, 'x': xy[:, 0], 'y': xy[:, 1], 'vol_peak': cross['vol_peak'].values,
                       'w': cross['streetwidth'].values, 's': cross['posted_speed'].values,
                       'l': cross['number_travel_lanes'].values})
    s = (cr.groupby('site')
         .agg(n_cross=('vol_peak', 'size'), site_vol=('vol_peak', 'sum'), max_width=('w', 'max'),
              max_speed=('s', 'max'), max_lanes=('l', 'max'), x=('x', 'mean'), y=('y', 'mean'))
         .reset_index())
    pts = gpd.GeoDataFrame(s[['site']], geometry=gpd.points_from_xy(s['x'], s['y']), crs=6538)
    j = gpd.sjoin(pts, tr, how='left', predicate='within')
    j = j[~j.index.duplicated(keep='first')]
    s['GEOID'] = j['GEOID'].values
    s = s.merge(tm0[cols], on='GEOID', how='inner')
    s['county'] = s['GEOID'].str[2:5].map(COUNTY_NAMES)
    pts = pts[pts['site'].isin(s['site'])]

    nearest = gpd.sjoin_nearest(cp, pts[['site', 'geometry']], how='left', max_distance=RADIUS_M)
    nearest = nearest[~nearest.index.duplicated(keep='first')]
    ok = nearest[nearest['site'].notna()]
    tot = ok.groupby('site').agg(harm=('site', 'size'), killed=('ped_kill', 'sum'))
    byyr = ok.groupby(['site', 'year']).size().unstack(fill_value=0)
    byyr.columns = [f'y{int(k)}' for k in byyr.columns]
    d = s.merge(tot, left_on='site', right_index=True, how='left').merge(byyr, left_on='site', right_index=True, how='left')
    ycols = [f'y{y}' for y in YEARS]
    for k in ['harm', 'killed'] + [k for k in ycols if k in d.columns]:
        d[k] = d[k].fillna(0)
    for k in ycols:
        if k not in d.columns:
            d[k] = 0
    early = [f'y{y}' for y in YEARS if y <= 2021]
    late = [f'y{y}' for y in YEARS if y >= 2022]
    d['early'] = d[early].sum(axis=1)
    d['late'] = d[late].sum(axis=1)
    d['log_vol'] = np.log(d['site_vol'].clip(lower=1))
    d['width10'] = d['max_width'] / 10
    d['speed5'] = d['max_speed'] / 5
    d['lanes'] = d['max_lanes']
    d = d.dropna(subset=['log_vol', 'width10', 'speed5', 'lanes', 'Poverty_Rate_Percent', 'Walk_Share',
                         'Transit_Share', 'Commercial_Share', 'ZCTA_cluster']).copy()
    for k in ['Walk_Share', 'Transit_Share', 'Commercial_Share', 'n_cross']:
        d[k + '_z'] = (d[k] - d[k].mean()) / d[k].std()
    return d, len(ok) / len(cp)


RHS = ('Poverty_Rate_Percent + log_vol + width10 + lanes + speed5 + n_cross_z + Walk_Share_z'
       ' + Transit_Share_z + Commercial_Share_z + C(county)')


def fit(d, y):
    groups = pd.factorize(d['ZCTA_cluster'])[0]
    return smf.glm(f'{y} ~ {RHS}', data=d, family=sm.families.Poisson()).fit(
        cov_type='cluster', cov_kwds={'groups': groups})


def rr(res):
    lo, hi = res.conf_int().loc['Poverty_Rate_Percent']
    return f"{np.exp(10 * res.params['Poverty_Rate_Percent']):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})"


# 1. Site definition
print('\n1. Site definition (30 m crash radius), model D')
datasets = {}
for eps in [15, 25, 40]:
    d, share = make_dataset(eps)
    datasets[eps] = d
    print(f'  grouping distance {eps} m: {len(d)} sites, {share:.1%} of crashes assigned, poverty RR per +10 = {rr(fit(d, "harm"))}')

# 2. Time
d = datasets[25]
print('\n2. Stability over time (same 2018 and 2019 volumes), 25 m sites')
for label, y in [('2019 to 2021', 'early'), ('2022 to 2023', 'late')] + [(str(yr), f'y{yr}') for yr in YEARS]:
    res = fit(d, y)
    print(f'  {label}: crashes = {int(d[y].sum())}, poverty RR per +10 = {rr(res)}, volume elasticity = {res.params["log_vol"]:.2f}')

# 3. Leave one borough out
print('\n3. Leave one borough out, 25 m sites')
for name in COUNTY_NAMES.values():
    sub = d[d['county'] != name].copy()
    print(f'  without {name}: {len(sub)} sites, poverty RR per +10 = {rr(fit(sub, "harm"))}')
````

````python
# Robustness battery for the pedestrian death gradient at crossing sites.
# Run in the TDSP Research Sandbox after the Milestone 1 and Milestone 5 cells.
# Uses these names from the session: crashes, tracts, tract_table, HEADLINE_YEARS.
# Allow 8 to 10 minutes. The first part rebuilds the crosswalks and sites.
#
# Checks
#  1. Model progression A to D for deaths, Poisson robust and negative binomial GLM.
#  2. Site grouping distance 15, 25, 40 m and crash radius 20, 30, 50 m.
#  3. Time: 2019 to 2021 versus 2022 to 2023, with death counts.
#  4. Leave one borough out.
#  5. Cluster bootstrap interval for the death ratio in model D.
#  6. Frequency versus severity: harm crashes per site, then deaths per harm crash.

# %% J1. Deaths battery
import os
import zipfile
import warnings
import numpy as np
import pandas as pd
import patsy
import geopandas as gpd
import statsmodels.api as sm
import statsmodels.formula.api as smf
from sklearn.cluster import DBSCAN
from google.colab import drive

warnings.filterwarnings('ignore')
drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
MEMBER = 'NYC_pednetwork_estimates_counts_2018-2019.geojson'
LOCAL = f'/content/{MEMBER}'
COUNTY_NAMES = {'005': 'Bronx', '047': 'Brooklyn', '061': 'Manhattan', '081': 'Queens', '085': 'Staten Island'}
YEARS = sorted(HEADLINE_YEARS)

if not os.path.exists(LOCAL):
    with zipfile.ZipFile(f'{DRIVE_DIR}/supp_data_1_network.zip') as z:
        with z.open(MEMBER) as src, open(LOCAL, 'wb') as dst:
            while True:
                chunk = src.read(1 << 20)
                if not chunk:
                    break
                dst.write(chunk)

# Crosswalks, built once
cl = gpd.read_file(f'{DRIVE_DIR}/centerline_inkn-q76z.geojson').to_crs(6538)
for c in ['rw_type', 'status', 'streetwidth', 'posted_speed', 'number_travel_lanes']:
    cl[c] = pd.to_numeric(cl[c], errors='coerce')
road = cl[(cl['status'] == 2) & (cl['rw_type'] == 1) & (cl['trafdir'] != 'NV')]
road = road[['streetwidth', 'posted_speed', 'number_travel_lanes', 'geometry']]

WKDY = ['predwkdyAM', 'predwkdyMD', 'predwkdyPM']
WKND = ['predwkndAM', 'predwkndMD', 'predwkndPM']
seg = gpd.read_file(LOCAL, engine='pyogrio', columns=WKDY + WKND).to_crs(6538)
seg[WKDY + WKND] = seg[WKDY + WKND].fillna(0).clip(lower=0)
seg['vol_peak'] = seg[WKDY + WKND].mean(axis=1)
seg['seg_id'] = np.arange(len(seg))
hit = gpd.sjoin(seg[['seg_id', 'geometry']], road, how='inner', predicate='intersects')
widest = (hit.sort_values('streetwidth', ascending=False, na_position='last')
          .drop_duplicates('seg_id')[['seg_id', 'streetwidth', 'posted_speed', 'number_travel_lanes']])
cross = seg.merge(widest, on='seg_id', how='inner')
mid = cross.geometry.interpolate(0.5, normalized=True)
xy = np.c_[mid.x.values, mid.y.values]
print('crosswalk segments:', len(cross))

tm0 = tract_table(HEADLINE_YEARS, 'local')
cols = ['GEOID', 'Poverty_Rate_Percent', 'Walk_Share', 'Transit_Share', 'Commercial_Share', 'ZCTA_cluster']
tr = tracts.to_crs(6538)[['GEOID', 'geometry']]

c = crashes[crashes['YEAR'].isin(HEADLINE_YEARS) & ~crashes['is_highway']].copy()
c['ped_inj'] = c['NUMBER OF PEDESTRIANS INJURED'].fillna(0)
c['ped_kill'] = c['NUMBER OF PEDESTRIANS KILLED'].fillna(0)
c = c[(c['ped_inj'] + c['ped_kill']) > 0]
c = c[c['LATITUDE'].between(40.4, 41.0) & c['LONGITUDE'].between(-74.3, -73.6)]
cp = gpd.GeoDataFrame(c[['ped_inj', 'ped_kill']].assign(year=c['YEAR'].values).reset_index(drop=True),
                      geometry=gpd.points_from_xy(c['LONGITUDE'], c['LATITUDE']), crs=4326).to_crs(6538)
print('pedestrian harm crashes:', len(cp), '| deaths among them:', int(cp['ped_kill'].sum()))


def make_dataset(eps, radius):
    lab = DBSCAN(eps=eps, min_samples=1).fit_predict(xy)
    cr = pd.DataFrame({'site': lab, 'x': xy[:, 0], 'y': xy[:, 1], 'vol_peak': cross['vol_peak'].values,
                       'w': cross['streetwidth'].values, 's': cross['posted_speed'].values,
                       'l': cross['number_travel_lanes'].values})
    s = (cr.groupby('site')
         .agg(n_cross=('vol_peak', 'size'), site_vol=('vol_peak', 'sum'), max_width=('w', 'max'),
              max_speed=('s', 'max'), max_lanes=('l', 'max'), x=('x', 'mean'), y=('y', 'mean'))
         .reset_index())
    pts = gpd.GeoDataFrame(s[['site']], geometry=gpd.points_from_xy(s['x'], s['y']), crs=6538)
    j = gpd.sjoin(pts, tr, how='left', predicate='within')
    j = j[~j.index.duplicated(keep='first')]
    s['GEOID'] = j['GEOID'].values
    s = s.merge(tm0[cols], on='GEOID', how='inner')
    s['county'] = s['GEOID'].str[2:5].map(COUNTY_NAMES)
    pts = pts[pts['site'].isin(s['site'])]

    near = gpd.sjoin_nearest(cp, pts[['site', 'geometry']], how='left', max_distance=radius)
    near = near[~near.index.duplicated(keep='first')]
    ok = near[near['site'].notna()]
    tot = ok.groupby('site').agg(harm=('site', 'size'), killed=('ped_kill', 'sum'))
    hy = ok.groupby(['site', 'year']).size().unstack(fill_value=0)
    hy.columns = [f'y{int(k)}' for k in hy.columns]
    ky = ok.groupby(['site', 'year'])['ped_kill'].sum().unstack(fill_value=0)
    ky.columns = [f'k{int(k)}' for k in ky.columns]
    d = (s.merge(tot, left_on='site', right_index=True, how='left')
          .merge(hy, left_on='site', right_index=True, how='left')
          .merge(ky, left_on='site', right_index=True, how='left'))
    for yr in YEARS:
        for pref in ['y', 'k']:
            if f'{pref}{yr}' not in d.columns:
                d[f'{pref}{yr}'] = 0
    allc = ['harm', 'killed'] + [f'y{yr}' for yr in YEARS] + [f'k{yr}' for yr in YEARS]
    d[allc] = d[allc].fillna(0)
    d['k_early'] = d[[f'k{yr}' for yr in YEARS if yr <= 2021]].sum(axis=1)
    d['k_late'] = d[[f'k{yr}' for yr in YEARS if yr >= 2022]].sum(axis=1)
    d['log_vol'] = np.log(d['site_vol'].clip(lower=1))
    d['width10'] = d['max_width'] / 10
    d['speed5'] = d['max_speed'] / 5
    d['lanes'] = d['max_lanes']
    d = d.dropna(subset=['log_vol', 'width10', 'speed5', 'lanes', 'Poverty_Rate_Percent', 'Walk_Share',
                         'Transit_Share', 'Commercial_Share', 'ZCTA_cluster']).copy()
    for k in ['Walk_Share', 'Transit_Share', 'Commercial_Share', 'n_cross']:
        d[k + '_z'] = (d[k] - d[k].mean()) / d[k].std()
    return d


DESIGN = 'log_vol + width10 + lanes + speed5'
CTRL = ' + n_cross_z + Walk_Share_z + Transit_Share_z + Commercial_Share_z'
SPECS = {'A volume only': 'log_vol', 'B + street design': DESIGN,
         'C + tract controls': DESIGN + CTRL, 'D + borough (primary)': DESIGN + CTRL + ' + C(county)'}
RHS_D = 'Poverty_Rate_Percent + ' + SPECS['D + borough (primary)']


def fit(d, y, rhs=RHS_D, offset=None):
    groups = pd.factorize(d['ZCTA_cluster'])[0]
    model = smf.glm(f'{y} ~ {rhs}', data=d, family=sm.families.Poisson(), offset=offset)
    return model.fit(cov_type='cluster', cov_kwds={'groups': groups})


def fit_nb(d, y, rhs):
    p = fit(d, y, rhs)
    mu = np.asarray(p.mu)
    yy = d[y].values
    alpha = max(float((mu * (((yy - mu) ** 2 - yy) / mu)).sum() / (mu * mu).sum()), 1e-6)
    groups = pd.factorize(d['ZCTA_cluster'])[0]
    model = smf.glm(f'{y} ~ {rhs}', data=d, family=sm.families.NegativeBinomial(alpha=alpha))
    return model.fit(cov_type='cluster', cov_kwds={'groups': groups}), alpha


def rr(res):
    lo, hi = res.conf_int().loc['Poverty_Rate_Percent']
    return f"{np.exp(10 * res.params['Poverty_Rate_Percent']):.2f} ({np.exp(10 * lo):.2f} to {np.exp(10 * hi):.2f})"


d25 = make_dataset(25, 30)
print(f'\nprimary dataset: {len(d25)} sites, {int(d25["harm"].sum())} harm crashes, {int(d25["killed"].sum())} deaths')

# 1. Model progression
print('\n1. Deaths, poverty rate ratio per +10 points')
for k, f in SPECS.items():
    rhs = 'Poverty_Rate_Percent + ' + f
    p = fit(d25, 'killed', rhs)
    nbres, alpha = fit_nb(d25, 'killed', rhs)
    print(f'  {k}: Poisson {rr(p)} | NB {rr(nbres)} (alpha {alpha:.2f})')

# 2. Site definition and radius
print('\n2. Site grouping distance and crash radius, model D')
for eps, radius in [(15, 30), (40, 30), (25, 20), (25, 50)]:
    dd = make_dataset(eps, radius)
    print(f'  grouping {eps} m, radius {radius} m: {int(dd["killed"].sum())} deaths, poverty RR = {rr(fit(dd, "killed"))}')

# 3. Time
print('\n3. Time, model D')
for label, y in [('2019 to 2021', 'k_early'), ('2022 to 2023', 'k_late')]:
    print(f'  {label}: {int(d25[y].sum())} deaths, poverty RR = {rr(fit(d25, y))}')
print('  deaths per year:', {yr: int(d25[f"k{yr}"].sum()) for yr in YEARS})

# 4. Leave one borough out
print('\n4. Leave one borough out, model D')
for name in COUNTY_NAMES.values():
    sub = d25[d25['county'] != name].copy()
    print(f'  without {name}: {int(sub["killed"].sum())} deaths, poverty RR = {rr(fit(sub, "killed"))}')


# 5. Cluster bootstrap for the death ratio
def irls(Xm, yv, beta0, iters=40, tol=1e-9):
    beta = beta0.copy()
    for _ in range(iters):
        eta = Xm @ beta
        mu = np.exp(np.clip(eta, -20, 20))
        z = eta + (yv - mu) / mu
        XtW = Xm.T * mu
        new = np.linalg.solve(XtW @ Xm, XtW @ z)
        done = np.max(np.abs(new - beta)) < tol
        beta = new
        if done:
            break
    return beta


yS, XS = patsy.dmatrices(f'killed ~ {RHS_D}', d25, return_type='dataframe')
Xv, yv = XS.values.astype(float), yS.iloc[:, 0].values.astype(float)
i_pov = list(XS.columns).index('Poverty_Rate_Percent')
cidx = pd.factorize(d25['ZCTA_cluster'])[0]
rows_by_cluster = [np.where(cidx == g)[0] for g in range(cidx.max() + 1)]
b0 = irls(Xv, yv, np.zeros(Xv.shape[1]), iters=80)
rng = np.random.default_rng(7)
boots = []
for _ in range(500):
    pick = rng.integers(0, len(rows_by_cluster), len(rows_by_cluster))
    idx = np.concatenate([rows_by_cluster[g] for g in pick])
    try:
        boots.append(irls(Xv[idx], yv[idx], b0)[i_pov])
    except np.linalg.LinAlgError:
        continue
lo, hi = np.percentile(np.exp(10 * np.array(boots)), [2.5, 97.5])
print(f'\n5. Cluster bootstrap, {len(boots)} resamples of {len(rows_by_cluster)} ZIP clusters, deaths, model D')
print(f'  point estimate {np.exp(10 * b0[i_pov]):.2f}, percentile interval {lo:.2f} to {hi:.2f}')

# 6. Frequency versus severity
freq = fit(d25, 'harm')
has = d25[d25['harm'] > 0].copy()
sev = fit(has, 'killed', offset=np.log(has['harm'].values))
dea = fit(d25, 'killed')
print('\n6. Frequency and severity, model D')
print(f'  harm crashes per site (frequency): {rr(freq)}')
print(f'  deaths per harm crash (severity, {len(has)} sites with a crash): {rr(sev)}')
print(f'  deaths per site (total): {rr(dea)}')

d25.to_csv(f'{DRIVE_DIR}/sites_eps25_r30_by_year.csv', index=False)
print('\nsaved to', f'{DRIVE_DIR}/sites_eps25_r30_by_year.csv')
````

#### <font color="blue">Step 10. Geographic scale and random zones</font>

````python
# Scale and zone effects on the poverty gradient.
# Question: how much does the poverty rate ratio depend on how the city is divided up?
#  - Scale effect: the same data aggregated to larger and larger zones.
#  - Zone effect: many random contiguous zonings of the same average size.
# Needs only crossing_sites_model_data.csv on Drive and the Census tract boundaries.
# No Milestone 1 or Milestone 5 cells are needed. Allow 2 to 4 minutes.
#
# Zones are built from the 2,100 or so tracts in the site table. Each zone sums crashes and crossing
# volume, and takes population weighted poverty, walking, transit and commercial shares.
# Two models are fit at every scale, residents only and residents plus crossing volume and street design.

# %% K1. Scale and zone effects
import os
import warnings
import numpy as np
import pandas as pd
import geopandas as gpd
import requests
import statsmodels.api as sm
import statsmodels.formula.api as smf
from google.colab import drive

warnings.filterwarnings('ignore')
drive.mount('/content/drive')
DRIVE_DIR = '/content/drive/MyDrive/TDSP/nyc_walks'
df = pd.read_csv(f'{DRIVE_DIR}/crossing_sites_model_data.csv', dtype={'GEOID': str})

df['w_x_vol'] = df['max_width'] * df['site_vol']
df['l_x_vol'] = df['max_lanes'] * df['site_vol']
df['s_x_vol'] = df['max_speed'] * df['site_vol']
t = (df.groupby('GEOID')
     .agg(harm=('harm_crashes', 'sum'), killed=('ped_killed', 'sum'), vol=('site_vol', 'sum'),
          wv=('w_x_vol', 'sum'), lv=('l_x_vol', 'sum'), sv=('s_x_vol', 'sum'),
          pop=('Total_Population', 'first'), pov=('Poverty_Rate_Percent', 'first'),
          walk=('Walk_Share', 'first'), transit=('Transit_Share', 'first'),
          comm=('Commercial_Share', 'first'), zip=('ZCTA_cluster', 'first'), county=('county', 'first'))
     .reset_index())
t = t[t['pop'] > 0].reset_index(drop=True)
print('tracts:', len(t), '| harm crashes:', int(t['harm'].sum()), '| deaths:', int(t['killed'].sum()))

# Tract adjacency (queen contiguity), within the same borough
TRACT_ZIP = '/content/ny_tracts2020.zip'
if not os.path.exists(TRACT_ZIP):
    url = 'https://www2.census.gov/geo/tiger/GENZ2020/shp/cb_2020_36_tract_500k.zip'
    with open(TRACT_ZIP, 'wb') as f:
        f.write(requests.get(url, timeout=120).content)
g = gpd.read_file(TRACT_ZIP)
g = g[g['GEOID'].isin(t['GEOID'])][['GEOID', 'geometry']].reset_index(drop=True)
pairs = gpd.sjoin(g, g, predicate='intersects')
pairs = pairs[pairs['GEOID_left'] != pairs['GEOID_right']]
county_of = dict(zip(t['GEOID'], t['county']))
pairs = pairs[pairs['GEOID_left'].map(county_of) == pairs['GEOID_right'].map(county_of)]
nbrs = pairs.groupby('GEOID_left')['GEOID_right'].apply(list).to_dict()
ids = list(t['GEOID'])


def random_zones(k, rng):
    zone_of = {}
    zid = 0
    for seed in rng.permutation(ids):
        if seed in zone_of:
            continue
        zone_of[seed] = zid
        size = 1
        frontier = [n for n in nbrs.get(seed, []) if n not in zone_of]
        while size < k and frontier:
            nxt = frontier.pop(int(rng.integers(len(frontier))))
            if nxt in zone_of:
                continue
            zone_of[nxt] = zid
            size += 1
            frontier += [n for n in nbrs.get(nxt, []) if n not in zone_of]
        zid += 1
    return t['GEOID'].map(zone_of).values


def aggregate(zone_ids):
    z = t.assign(zone=zone_ids)
    for c in ['pov', 'walk', 'transit', 'comm']:
        z[c + '_p'] = z[c] * z['pop']
    grp = z.groupby('zone')
    out = grp[['harm', 'killed', 'vol', 'wv', 'lv', 'sv', 'pop', 'pov_p', 'walk_p', 'transit_p', 'comm_p']].sum()
    for c in ['pov', 'walk', 'transit', 'comm']:
        out[c] = out[c + '_p'] / out['pop']
    out['width10'] = out['wv'] / out['vol'] / 10
    out['lanes'] = out['lv'] / out['vol']
    out['speed5'] = out['sv'] / out['vol'] / 5
    out['log_pop'] = np.log(out['pop'])
    out['log_vol'] = np.log(out['vol'].clip(lower=1))
    for c in ['walk', 'transit', 'comm']:
        out[c + '_z'] = (out[c] - out[c].mean()) / out[c].std()
    idx = z.groupby('zone')['pop'].idxmax()
    out['county'] = z.loc[idx.values, 'county'].values
    return out.reset_index()


FORMS = {
    'residents only': 'pov + log_pop + walk_z + transit_z + comm_z + C(county)',
    'residents + volume + design': 'pov + log_pop + log_vol + width10 + lanes + speed5 + walk_z + transit_z + comm_z + C(county)',
}


def fit_rr(zdf, y, form, groups=None):
    kw = {'cov_type': 'HC1'} if groups is None else {'cov_type': 'cluster', 'cov_kwds': {'groups': groups}}
    res = smf.glm(f'{y} ~ {form}', data=zdf, family=sm.families.Poisson()).fit(**kw)
    lo, hi = res.conf_int().loc['pov']
    return np.exp(10 * res.params['pov']), np.exp(10 * lo), np.exp(10 * hi)


def show(label, rrs):
    r, lo, hi = rrs
    return f'{label}: {r:.2f} ({lo:.2f} to {hi:.2f})'


tract_z = aggregate(t['GEOID'].values)
zip_z = aggregate(t['zip'].values)
zip_groups = pd.factorize(t['zip'])[0]
print(f'ZIP area zones: {len(zip_z)}')

rng = np.random.default_rng(11)
SIZES = [2, 4, 8, 16, 32]
R = 100
for y, name in [('harm', 'Pedestrian harm crashes'), ('killed', 'Pedestrian deaths')]:
    print(f'\n=== {name}: poverty rate ratio per +10 points ===')
    for fname, form in FORMS.items():
        print(f'\n  Model: {fname}')
        print('   ', show(f'tracts ({len(tract_z)} zones, ZIP clustered)', fit_rr(tract_z, y, form, zip_groups)))
        print('   ', show(f'ZIP areas ({len(zip_z)} zones)', fit_rr(zip_z, y, form)))
    # Random zonings, same set of zonings for both models
print('\n=== Random contiguous zonings, point estimates across', R, 'draws ===')
store = {}
for k in SIZES:
    zonings = [random_zones(k, rng) for _ in range(R)]
    store[k] = zonings
    zsizes = [len(np.unique(z)) for z in zonings]
    print(f'\n  target {k} tracts per zone: about {np.mean(zsizes):.0f} zones per draw, mean size {len(t) / np.mean(zsizes):.1f} tracts')
    for y, name in [('harm', 'harm crashes'), ('killed', 'deaths')]:
        for fname, form in FORMS.items():
            vals = []
            for z in zonings:
                try:
                    vals.append(fit_rr(aggregate(z), y, form)[0])
                except Exception:
                    continue
            vals = np.array(vals)
            lo, med, hi = np.percentile(vals, [2.5, 50, 97.5])
            print(f'    {name}, {fname}: median {med:.2f}, 95% of zonings between {lo:.2f} and {hi:.2f}')
print('\nReference from earlier runs, site level model D: harm crashes 1.16, deaths 1.29')
````
